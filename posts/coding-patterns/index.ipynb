**General Notes on Patterns:**

*   **Patterns are Solutions to Recurring Problems:** These aren't just arbitrary techniques; they represent proven solutions to frequently encountered problems.  Knowing them allows you to quickly identify the *type* of problem and apply a known-good approach.
*   **Understanding *Why* is Crucial:** Don't just memorize the code. Understand *why* a pattern works, its time and space complexity implications, and when it's the *best* choice (versus alternatives).
*   **Adapt, Don't Copy-Paste:**  Interview problems will rarely be identical to textbook examples.  You'll need to adapt the core pattern to the specific constraints and requirements of the problem.
* **Naming Consistency**: There is a common list of named patterns, but some interviewers or companies will use different names. Make sure to understand the pattern conceptually so the name is not important.
* **Categorization**: I have grouped patterns where helpful.

**1. Two Pointers**

*   **Description:** Uses two pointers (usually indices or iterators) to traverse a data structure (often an array or linked list) simultaneously.  The pointers can move in the same direction, opposite directions, or at different speeds.
*   **Identification:** Problems involving finding pairs, subarrays, or sequences that meet a specific condition, often in a *sorted* array or list. Look for phrases like "closest pair," "subarray sum," "two elements that add up to," or "remove duplicates."
*   **Example:** Find two numbers in a sorted array that add up to a target.

```python
def two_sum(nums, target):
    left = 0
    right = len(nums) - 1

    while left < right:
        current_sum = nums[left] + nums[right]
        if current_sum == target:
            return [left, right]  # Or return the values: [nums[left], nums[right]]
        elif current_sum < target:
            left += 1
        else:
            right -= 1

    return None  # No solution found
```

```python
# Example Usage and Output
nums1 = [2, 7, 11, 15]
target1 = 9
result1 = two_sum(nums1, target1)
print(f"Input: nums = {nums1}, target = {target1}, Output: {result1}")
```

```python
nums2 = [2, 3, 4]
target2 = 6
result2 = two_sum(nums2, target2)
print(f"Input: nums = {nums2}, target = {target2}, Output: {result2}")

```

*   **Alternatives:** Hash maps (for unsorted arrays, see below), brute-force (nested loops - inefficient).
*   **Complexity:** Often O(n) time, O(1) space (if modifying in place).  Brute force is O(n^2).


**2. Fast & Slow Pointers (Floyd's Cycle Detection)**

*   **Description:** A specialized form of the Two Pointers pattern where one pointer (fast) moves faster (e.g., two steps at a time) than the other (slow).  Primarily used to detect cycles in linked lists or find the middle of a linked list.
*   **Identification:** Problems involving linked lists, cycle detection, finding the middle element, or determining if a linked list has a loop.
*   **Example:** Detect if a linked list has a cycle.

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def has_cycle(head: ListNode) -> bool:
    slow = head
    fast = head

    while fast is not None and fast.next is not None:
        slow = slow.next
        fast = fast.next.next
        if slow == fast:
            return True  # Cycle detected

    return False  # No cycle
```

*   **Alternatives:** Using a hash set to store visited nodes (requires O(n) space).
*   **Complexity:** O(n) time, O(1) space.

**3. Sliding Window**

*   **Description:** Maintains a "window" (a contiguous subarray or substring) that slides over the input. The window size can be fixed or variable. Useful for problems involving subarrays/substrings that meet certain criteria.
*   **Identification:** Problems asking for the maximum/minimum/longest/shortest subarray/substring that satisfies a condition (e.g., "longest substring without repeating characters," "maximum sum subarray of size k").
*   **Example:** Find the maximum sum of a subarray of size `k`.


```python
def max_sum_subarray(nums, k):
    if len(nums) < k:
        return None  # Or raise an exception

    window_sum = sum(nums[:k])
    max_sum = window_sum

    for i in range(k, len(nums)):
        window_sum += nums[i] - nums[i - k]  # Slide the window
        max_sum = max(max_sum, window_sum)

    return max_sum
```

*   **Alternatives:** Brute force (check all possible subarrays - O(n^2) or worse).
*   **Complexity:** Often O(n) time, O(1) space (for fixed-size windows).  Variable-size windows can sometimes be O(n) but require careful analysis.

**4. Merge Intervals**

*   **Description:** Deals with overlapping intervals. The core idea is to sort the intervals and then merge those that overlap.
*   **Identification:** Problems explicitly mentioning intervals (start and end times/positions), and asking to merge overlapping ones, find non-overlapping intervals, or insert a new interval.
*   **Example:** Merge overlapping intervals.

```python
def merge_intervals(intervals):
    intervals.sort(key=lambda x: x[0])  # Sort by start time
    merged = []
    for interval in intervals:
        # If the list of merged intervals is empty or if the current
        # interval does not overlap with the previous, simply append it.
        if not merged or merged[-1][1] < interval[0]:
            merged.append(interval)
        else:
            # Otherwise, there is overlap, so we merge the current and previous intervals.
            merged[-1][1] = max(merged[-1][1], interval[1])
    return merged
```

*   **Alternatives:**  Brute force (compare every interval pair - O(n^2)).
*   **Complexity:** O(n log n) for sorting, then O(n) for merging. Overall, O(n log n).

**5. Cyclic Sort**

*   **Description:** Used when dealing with arrays containing numbers in a specific *range* (e.g., 1 to n).  The idea is to place each number at its correct index (e.g., number `3` should be at index `2`).
*   **Identification:** Problems involving a sorted array with numbers in a known range, often missing one or more numbers, or finding duplicates within that range. The array is often *not* sorted initially, but the range is known.
*   **Example:** Find the missing number in an array containing numbers from 1 to n (one number missing).

```python
def find_missing_number(nums):
    i = 0
    while i < len(nums):
        correct_index = nums[i] - 1  # Calculate the correct index for the number
        if nums[i] != nums[correct_index]:
            nums[i], nums[correct_index] = nums[correct_index], nums[i]  # Swap
        else:
            i += 1

    # Now, find the missing number (the one not in its correct place)
    for j in range(len(nums)):
        if nums[j] != j + 1:
            return j + 1

    return len(nums) + 1 # If all numbers 1..n are present, n+1 is missing.
```

*   **Alternatives:** Hash set (requires O(n) space), sum formula (only works for finding a single missing number).
*   **Complexity:** O(n) time, O(1) space (in-place).

**6. In-place Reversal of a Linked List**

*   **Description:** Reverses a linked list by modifying the `next` pointers of the nodes, without using extra space (beyond a few temporary variables).
*   **Identification:** Problems explicitly asking to reverse a linked list, or a portion of it.
*   **Example:** Reverse a linked list.

```python
def reverse_linked_list(head):
    prev = None
    current = head
    while current:
        next_node = current.next  # Store the next node
        current.next = prev      # Reverse the pointer
        prev = current           # Move prev forward
        current = next_node      # Move current forward
    return prev  # prev is the new head
```

*   **Alternatives:** Recursive approach (uses call stack space, so not strictly in-place).
*   **Complexity:** O(n) time, O(1) space.


**7. Stacks**

* **Description:** A stack is a linear data structure that follows the Last-In, First-Out (LIFO) principle. Elements are added (pushed) and removed (popped) from the top of the stack.
* **Identification:** Problems involving:
    *   Reversing order of elements.
    *   Matching parentheses, brackets, or braces.
    *   Evaluating expressions (infix, postfix).
    *   Undo/redo functionality.
    *   Function call management (call stack).
    *   Depth-First Search (DFS) - see below (iterative DFS uses a stack).
* **Example:** Check if a string of parentheses is balanced.

```python
def is_balanced(s):
    stack = []
    mapping = {')': '(', '}': '{', ']': '['}
    for char in s:
        if char in mapping:
            top_element = stack.pop() if stack else '#'  # Handle empty stack
            if mapping[char] != top_element:
                return False
        else:
            stack.append(char)
    return not stack  # Stack should be empty if balanced
```

* **Alternatives:** Sometimes recursion (which implicitly uses the call stack).
* **Complexity:** Push and pop operations are O(1).

**8. Monotonic Stack**

*   **Description:** A stack where the elements are always in increasing or decreasing order (from bottom to top).  Useful for finding the next greater/smaller element in an array.
*   **Identification:** Problems asking for the "next greater element," "next smaller element," "previous greater element," or similar variations.  Also look for problems where you need to maintain a sorted order within a sliding window.
*   **Example:** Find the next greater element for each element in an array.

```python

def next_greater_element(nums):
    result = [-1] * len(nums)
    stack = []  # Store indices

    for i in range(len(nums)):
        while stack and nums[stack[-1]] < nums[i]:
            index = stack.pop()
            result[index] = nums[i]
        stack.append(i)
    return result
```


*   **Alternatives:** Brute force (nested loops - O(n^2)).
*   **Complexity:** O(n) time, O(n) space (in the worst case, all elements are in the stack).

**9. Hash Maps (Dictionaries in Python)**

*   **Description:** Stores key-value pairs, allowing for fast lookups (average O(1)) based on keys.  Keys must be unique and hashable (immutable).
*   **Identification:** Problems requiring:
    *   Counting the frequency of elements.
    *   Checking for the existence of an element quickly.
    *   Storing and retrieving data associated with a specific key.
    *   Implementing caches.
    *   Finding pairs that sum to a target (unsorted array version of Two Pointers).
*   **Example:** Count the frequency of each character in a string.


```python
def character_frequency(s):
    counts = {}  # Dictionary to store counts
    for char in s:
        counts[char] = counts.get(char, 0) + 1
    return counts
```

*   **Alternatives:**  Arrays (if keys are integers and the range is small), sorted arrays + binary search (for sorted data).
*   **Complexity:** Average case: O(1) for insertion, deletion, and lookup. Worst case (hash collisions): O(n).

## **--- Tree and Graph Traversal ---**

These are fundamental algorithms, often used as building blocks for other patterns.

**10. Tree Breadth-First Search (BFS)**

*   **Description:** Traverses a tree level by level, visiting all nodes at a given depth before moving to the next depth.  Uses a queue.
*   **Identification:** Problems requiring level-order traversal, finding the shortest path in an unweighted graph, or processing nodes in a breadth-wise manner.
*   **Example:** Level-order traversal of a binary tree.

```python
from collections import deque

class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

def level_order(root):
    if not root:
        return []

    result = []
    queue = deque([root])

    while queue:
        level_size = len(queue)
        current_level = []
        for _ in range(level_size):
            node = queue.popleft()
            current_level.append(node.val)
            if node.left:
                queue.append(node.left)
            if node.right:
                queue.append(node.right)
        result.append(current_level)

    return result
```

*   **Alternatives:** Depth-First Search (if order doesn't matter or you need to go deep first).
*   **Complexity:** O(V + E) time (where V is the number of vertices/nodes and E is the number of edges), O(W) space (where W is the maximum width of the tree).

**11. Tree Depth-First Search (DFS)**

*   **Description:** Traverses a tree by going as deep as possible along each branch before backtracking.  Can be implemented recursively or iteratively (using a stack).  There are three main types: pre-order, in-order, and post-order.
*   **Identification:** Problems involving searching for a node, checking properties of paths, or processing nodes in a depth-wise manner.  Often used for problems on binary search trees (BSTs).
*   **Example:** Pre-order traversal (recursive).

```python
def pre_order_traversal(root):
    if not root:
        return []

    result = [root.val]  # Visit the root
    result.extend(pre_order_traversal(root.left))  # Traverse left subtree
    result.extend(pre_order_traversal(root.right))  # Traverse right subtree
    return result
```

*   **Alternatives:** Breadth-First Search (if level-order is needed).
*   **Complexity:** O(V + E) time, O(H) space in the average case (where H is the height of the tree - balanced tree), O(N) space in the worst case (skewed tree).

**12. Graphs**

*   **Description:** A data structure consisting of vertices (nodes) and edges connecting them. Graphs can be directed or undirected, weighted or unweighted, cyclic or acyclic.
*   **Identification:** Problems involving relationships between objects, networks, paths, connectivity, or shortest paths. Key terms: "nodes," "edges," "connections," "paths," "cycles," "neighbors."
*   **Representations:**
    *   **Adjacency Matrix:**  A 2D array where `matrix[i][j] = 1` if there's an edge from node i to node j (or the weight of the edge).  Good for dense graphs.
    *   **Adjacency List:** A dictionary (or array) where each key is a node and the value is a list of its neighbors.  Good for sparse graphs.
*   **Example:**  BFS for finding the shortest path in an unweighted graph (using adjacency list representation).

```python
from collections import deque

def shortest_path(graph, start, end):
    queue = deque([(start, [start])])  # (node, path)
    visited = {start}

    while queue:
        node, path = queue.popleft()
        if node == end:
            return path

        for neighbor in graph[node]:
            if neighbor not in visited:
                visited.add(neighbor)
                queue.append((neighbor, path + [neighbor]))

    return None  # No path found
```

*   **Algorithms:**
    *   **BFS (Breadth-First Search):** Shortest path in unweighted graphs.
    *   **DFS (Depth-First Search):**  Finding connected components, detecting cycles.
    *   **Dijkstra's Algorithm:** Shortest path in weighted graphs (non-negative weights).
    *   **Bellman-Ford Algorithm:** Shortest path in weighted graphs (can handle negative weights).
    *   **Topological Sort:** Ordering nodes in a directed acyclic graph (DAG).
    *   **Minimum Spanning Tree (MST) algorithms (Prim's, Kruskal's):** Finding a tree that connects all nodes with minimum total edge weight.
*   **Complexity:** Varies depending on the algorithm and graph representation. BFS and DFS are typically O(V + E).

**13. Island (Matrix Traversal)**

*   **Description:** Problems involving a 2D grid (matrix) where you need to find "islands" (connected regions of '1's or some other value) or perform operations on them.  Essentially a specialized graph traversal on a grid.
*   **Identification:** Problems with a 2D grid, often using terms like "island," "connected components," "region," or "area."
*   **Example:** Count the number of islands in a 2D grid (where '1' represents land and '0' represents water).

## **--- Heap-Based Patterns ---**

**14. Two Heaps**

*   **Description:** Uses two heaps (a min-heap and a max-heap) to maintain a set of numbers, allowing you to efficiently find the median (or other percentiles) of the set.
*   **Identification:** Problems asking for the running median of a stream of numbers, or finding elements closest to the median.
*   **Example:** Find the median of a stream of numbers.

```python
import heapq

class MedianFinder:
    def __init__(self):
        self.max_heap = []  # Stores the smaller half (negated for max-heap)
        self.min_heap = []  # Stores the larger half

    def add_num(self, num):
        if not self.max_heap or num <= -self.max_heap[0]:
            heapq.heappush(self.max_heap, -num)
        else:
            heapq.heappush(self.min_heap, num)

        # Balance the heaps
        if len(self.max_heap) > len(self.min_heap) + 1:
            heapq.heappush(self.min_heap, -heapq.heappop(self.max_heap))
        elif len(self.min_heap) > len(self.max_heap):
            heapq.heappush(self.max_heap, -heapq.heappop(self.min_heap))

    def find_median(self):
        if len(self.max_heap) == len(self.min_heap):
            return (-self.max_heap[0] + self.min_heap[0]) / 2.0
        else:
            return -self.max_heap[0]
```


*   **Alternatives:**  Sorted list (inefficient for insertions), balanced binary search trees (more complex to implement).
*   **Complexity:** `add_num`: O(log n), `find_median`: O(1).

**16. K-way Merge**

*   **Description:** Merging k sorted arrays (or linked lists) into a single sorted array.  Often uses a min-heap.
*   **Identification:** Problems explicitly mentioning merging multiple *sorted* lists or arrays.
*   **Example:** Merge k sorted linked lists.


```python
import heapq

class ListNode:  # Assuming ListNode definition is available
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def merge_k_sorted_lists(lists):
    min_heap = []
    # Add the first element from each list to the heap (along with list index)
    for i, lst in enumerate(lists):
        if lst:
            heapq.heappush(min_heap, (lst.val, i, lst))  # (value, list_index, node)

    dummy_head = ListNode()
    tail = dummy_head

    while min_heap:
        val, list_index, node = heapq.heappop(min_heap)
        tail.next = node
        tail = tail.next

        if node.next:
            heapq.heappush(min_heap, (node.next.val, list_index, node.next))

    return dummy_head.next
```

*   **Alternatives:**  Merge two lists at a time repeatedly (less efficient).
*   **Complexity:** O(N log k) time (where N is the total number of elements and k is the number of lists), O(k) space (for the heap).

## **--- Other Important Patterns ---**
**17. Subsets**

* **Description**: Generating all possible subsets (or power set) of a given set.
*   **Identification:** Problems asking for all combinations, all possible subsets, or power set.
* **Example**: Generate all subsets of a set.

```python
def subsets(nums):
    result = [[]]
    for num in nums:
        result += [curr + [num] for curr in result]
    return result
```

*   **Alternatives:** Recursive approach (backtracking - see below), bit manipulation.
*   **Complexity:** O(2^n) time and space (since there are 2^n subsets).

**18. Modified Binary Search**

* **Description:** Adapting binary search to solve problems beyond simply finding an element in a sorted array. The key is to modify the condition for moving the left and right pointers.
*   **Identification:** Problems involving sorted arrays where you need to find:
    *   The first/last occurrence of an element.
    *   The smallest/largest element that satisfies a condition.
    *   A peak element (element greater than its neighbors).
    *   The closest element to a target.
    *   The insertion point for a new element.
* **Example:** Find the first occurrence of a target in a sorted array.

```python
def first_occurrence(nums, target):
    left, right = 0, len(nums) - 1
    result = -1
    while left <= right:
        mid = (left + right) // 2
        if nums[mid] == target:
            result = mid
            right = mid - 1  # Keep searching on the left
        elif nums[mid] < target:
            left = mid + 1
        else:
            right = mid - 1
    return result
```

*   **Alternatives:** Linear search (O(n)).
*   **Complexity:** O(log n) time.

**19. Bitwise XOR**

* **Description:** Using the XOR (^) operator for bit manipulation. Key properties:
    *   `x ^ 0 = x`
    *   `x ^ x = 0`
    *   `x ^ y = y ^ x` (commutative)
    *   `(x ^ y) ^ z = x ^ (y ^ z)` (associative)
* **Identification:** Problems involving:
    *   Finding the single number that appears an odd number of times in an array (where all other numbers appear an even number of times).
    *   Swapping two numbers without a temporary variable.
    *   Detecting if two integers have opposite signs.
* **Example:** Find the single number that appears once in an array where all other numbers appear twice.


```python
def single_number(nums):
    result = 0
    for num in nums:
        result ^= num
    return result
```

*  **Alternatives:** Hash map (for counting frequencies - requires O(n) space).
* **Complexity:** O(n) time, O(1) space (for the single number problem).

**20. 0/1 Knapsack (Dynamic Programming)**

*   **Description:** A classic dynamic programming problem. You have a knapsack with a maximum weight capacity, and a set of items, each with a weight and a value.  The goal is to choose a subset of items to maximize the total value without exceeding the capacity.  "0/1" means you either take an item completely or don't take it at all (no fractions).
*   **Identification:** Problems involving choosing items with weights and values to maximize value within a capacity constraint.  Look for phrases like "maximum value," "capacity," "weights," and "values."
*   **Example:** Solve the 0/1 knapsack problem.

```python
def knapsack(weights, values, capacity):
    n = len(weights)
    dp = [[0] * (capacity + 1) for _ in range(n + 1)]

    for i in range(1, n + 1):
        for w in range(1, capacity + 1):
            if weights[i - 1] <= w:
                dp[i][w] = max(values[i - 1] + dp[i - 1][w - weights[i - 1]], dp[i - 1][w])
            else:
                dp[i][w] = dp[i - 1][w]

    return dp[n][capacity]
```

*   **Alternatives:**  Recursive approach (exponential time complexity), greedy approach (doesn't always work for 0/1 knapsack).
*   **Complexity:** O(n * capacity) time and space (where n is the number of items).

**21. Backtracking**

*   **Description:** A general algorithmic technique for finding all (or some) solutions to a computational problem, particularly constraint satisfaction problems.  It incrementally builds candidates to the solutions and abandons a candidate ("backtracks") as soon as it determines that the candidate cannot possibly be completed to a valid solution.
*   **Identification:** Problems requiring finding *all* possible solutions that satisfy certain constraints.  Commonly used for:
    *   Generating permutations and combinations.
    *   Solving Sudoku.
    *   N-Queens problem.
    *   Constraint satisfaction problems.
*   **Example:** Generate all permutations of a list of numbers.


```python
def permute(nums):
    result = []
    def backtrack(index, current_permutation):
        if index == len(nums):
            result.append(current_permutation[:])  # Add a copy
            return

        for i in range(len(current_permutation) + 1):
            current_permutation.insert(i, nums[index]) #insert number
            backtrack(index + 1, current_permutation)
            current_permutation.pop(i) #remove for other possibilities

    backtrack(0, [])
    return result
```

*   **Alternatives:**  Iterative approaches (often less readable), dynamic programming (for optimization problems).
*   **Complexity:**  Varies greatly depending on the problem. Often exponential (e.g., O(n!) for permutations).

**22. Trie (Prefix Tree)**

*   **Description:** A tree-like data structure used to store a dynamic set or associative array where the keys are usually strings.  Each node represents a character, and the path from the root to a node represents a prefix.
*   **Identification:** Problems involving:
    *   Efficient prefix searching (autocomplete).
    *   Storing and searching for words in a dictionary.
    *   Implementing spell checkers.
*   **Example:** Implement a Trie with `insert`, `search`, and `startsWith` methods.


```python
class TrieNode:
    def __init__(self):
        self.children = {}  # Map character to TrieNode
        self.is_end_of_word = False

class Trie:
    def __init__(self):
        self.root = TrieNode()

    def insert(self, word):
        node = self.root
        for char in word:
            if char not in node.children:
                node.children[char] = TrieNode()
            node = node.children[char]
        node.is_end_of_word = True

    def search(self, word):
        node = self.root
        for char in word:
            if char not in node.children:
                return False
            node = node.children[char]
        return node.is_end_of_word

    def startsWith(self, prefix):
        node = self.root
        for char in prefix:
            if char not in node.children:
                return False
            node = node.children[char]
        return True
```

*   **Alternatives:** Hash maps (for exact word matching, but not prefix searching), sorted arrays + binary search (for prefix searching, but less efficient for insertions/deletions).
*   **Complexity:** `insert`, `search`, and `startsWith`: O(m) time (where m is the length of the word/prefix). Space complexity depends on the number of words and their shared prefixes.

**23. Topological Sort (Graph)**

*   **Description:**  A linear ordering of vertices in a directed acyclic graph (DAG) such that for every directed edge from vertex u to vertex v, vertex u comes before vertex v in the ordering.
*   **Identification:** Problems involving dependencies between tasks, scheduling, or any scenario where you have a DAG and need to find a valid order of execution.
*   **Example:** Perform a topological sort on a DAG.

```python
from collections import defaultdict, deque

def topological_sort(graph):
    in_degree = defaultdict(int)
    for node in graph:
        for neighbor in graph[node]:
            in_degree[neighbor] += 1

    queue = deque([node for node in graph if in_degree[node] == 0])
    result = []

    while queue:
        node = queue.popleft()
        result.append(node)

        for neighbor in graph[node]:
            in_degree[neighbor] -= 1
            if in_degree[neighbor] == 0:
                queue.append(neighbor)

    if len(result) != len(graph):
        return None  # Cycle detected (not a DAG)
    return result
```

*   **Alternatives:**  Depth-First Search (another way to implement topological sort).
*   **Complexity:** O(V + E) time and space.

**24. Union Find (Disjoint Set)**

*   **Description:** A data structure that keeps track of a set of elements partitioned into a number of disjoint (non-overlapping) subsets.  It provides two main operations:
    *   `find(element)`:  Determines which subset an element belongs to.
    *   `union(element1, element2)`:  Merges the subsets containing `element1` and `element2`.
*   **Identification:** Problems involving:
    *   Detecting cycles in an undirected graph.
    *   Finding connected components in a graph.
    *   Maintaining groups of connected objects.
*   **Example:** Implement Union-Find with path compression and union by rank.

```python
class UnionFind:
    def __init__(self, n):
        self.parent = list(range(n))
        self.rank = [0] * n  # Used for union by rank

    def find(self, x):
        if self.parent[x] != x:
            self.parent[x] = self.find(self.parent[x])  # Path compression
        return self.parent[x]

    def union(self, x, y):
        root_x = self.find(x)
        root_y = self.find(y)
        if root_x != root_y:
            if self.rank[root_x] < self.rank[root_y]:
                self.parent[root_x] = root_y
            elif self.rank[root_x] > self.rank[root_y]:
                self.parent[root_y] = root_x
            else:
                self.parent[root_y] = root_x
                self.rank[root_x] += 1
                
```

*   **Alternatives:**  Depth-First Search or Breadth-First Search (for finding connected components, but less efficient for repeated union operations).
*   **Complexity:**  With path compression and union by rank, `find` and `union` operations have amortized time complexity close to O(α(n)), where α(n) is the inverse Ackermann function, which grows extremely slowly (almost constant for practical input sizes).

**25. Ordered Set**

*   **Description:**  A data structure that maintains a set of elements in sorted order, supporting efficient insertion, deletion, searching, and finding the k-th smallest element. In Python, you often achieve this functionality using a combination of data structures, or through external libraries, since Python's built-in `set` is *not* ordered.
*   **Identification:** Problems where you need a combination of set-like behavior (uniqueness) and sorted order operations:
    *   Finding the rank of an element (how many elements are smaller than it).
    *   Finding the k-th smallest element.
    *   Maintaining a sorted set with frequent insertions/deletions.
*   **Example (using `sortedcontainers` library - a common and efficient choice):**

```python
# Install: pip install sortedcontainers
from sortedcontainers import SortedSet

def ordered_set_example():
    s = SortedSet()
    s.add(5)
    s.add(2)
    s.add(8)
    s.add(2)  # Duplicate is ignored

    print(s)  # Output: SortedSet([2, 5, 8])

    print(s.bisect_left(5))  # Output: 1 (index where 5 would be inserted)
    print(s[1]) #Output 5 (element at index 1)
    print(s.index(5)) #Output 1 (index of element 5)

    s.remove(5)
    print(s)  # Output: SortedSet([2, 8])
```

*   **Alternatives:**
    *   **`list` + `bisect`:**  You *can* maintain a sorted list using the `bisect` module for insertion, but removing elements is O(n), making it inefficient for frequent deletions.
    *   **Balanced Binary Search Trees (e.g., Red-Black Trees):**  These provide the necessary operations, but are complex to implement from scratch in an interview.  `sortedcontainers` uses a highly optimized implementation of these.
*   **Complexity (using `sortedcontainers`):** Most operations (add, remove, search, bisect_left/right) are O(log n).  Finding the k-th element (`s[k]`) is also O(log n).

**26. Prefix Sum**

*   **Description:**  Preprocessing an array (or list) to create a new array where each element at index `i` stores the sum of all elements up to (and including) index `i` in the original array. This allows for fast calculation of the sum of any subarray.
*   **Identification:** Problems where you need to repeatedly calculate the sum of subarrays within a given array.  Look for phrases like "sum of subarray," "range sum query," or similar.
*   **Example:** Calculate the sum of a subarray from index `i` to `j` (inclusive) efficiently.

```python
def prefix_sum(nums):
    prefix = [0] * (len(nums) + 1)  # One extra element for convenience
    for i in range(len(nums)):
        prefix[i + 1] = prefix[i] + nums[i]
    return prefix

def subarray_sum(prefix, i, j):
    return prefix[j + 1] - prefix[i]

# Example usage
nums = [1, 2, 3, 4, 5]
prefix = prefix_sum(nums)
print(subarray_sum(prefix, 1, 3))  # Sum of elements from index 1 to 3 (2 + 3 + 4 = 9)
```

*   **Alternatives:**  Iterating through the subarray each time (O(n) per query).
*   **Complexity:** Building the prefix sum array: O(n).  Querying the sum of a subarray: O(1). This is a huge improvement over repeated iteration.

**27. Multi-threaded**

*   **Description:**  Using multiple threads of execution to perform tasks concurrently, potentially speeding up computations. Python's `threading` module is used for this.  *Crucially*, due to Python's Global Interpreter Lock (GIL), threads in CPython (the standard Python implementation) don't achieve true parallelism for CPU-bound tasks; they are better suited for I/O-bound tasks (waiting for network requests, disk operations, etc.).
* **Identification:**
    * Problems explicitly mentioning concurrency or parallelism.
    * Problems with independent tasks that can be executed concurrently.
    * Problems that involve waiting for external resources (network, I/O) where threads can release the GIL and allow other threads to run.
    * Problems that specify the performance characteristics want multithreading.
* **Example:**  Fetch multiple URLs concurrently.


```python
import threading
import requests  # You'll need to install this: pip install requests
import time

def fetch_url(url, results):
    try:
        response = requests.get(url)
        response.raise_for_status() # Raise HTTPError for bad requests (4xx or 5xx)
        results[url] = response.text  # Or process the response as needed
    except requests.exceptions.RequestException as e:
        results[url] = f"Error: {e}"

def multi_threaded_fetch(urls):
    threads = []
    results = {}  # Use a dictionary to store results with URLs as keys
    start_time = time.time()

    for url in urls:
        thread = threading.Thread(target=fetch_url, args=(url, results))
        threads.append(thread)
        thread.start()

    for thread in threads:
        thread.join()  # Wait for all threads to finish
    end_time = time.time()
    print(f"Fetched {len(urls)} URLs in {end_time - start_time:.2f} seconds")

    return results

# Example usage
urls = [
    "https://www.google.com",
    "https://www.wikipedia.org",
    "https://www.example.com",
]
results = multi_threaded_fetch(urls)
#for url, content in results.items(): #uncomment to print to standard output
#    print(f"--- {url} ---")
#    print(content[:200])  # Print the first 200 characters of each result
```

*   **Alternatives:**
    *   **`multiprocessing`:** For CPU-bound tasks, use the `multiprocessing` module, which bypasses the GIL by creating separate processes (each with its own interpreter).
    *   **`asyncio`:** For I/O-bound tasks, `asyncio` provides an asynchronous, single-threaded concurrency model that can be very efficient (and often preferred over threads).
    *   **Sequential processing:**  The simplest approach, but potentially slow for I/O-bound or independent tasks.
*   **Complexity:**  Complexity is highly dependent on the task.  The main benefit is reduced *wall-clock time* (the actual time it takes to complete), especially for I/O-bound tasks.  CPU-bound tasks *may* see some improvement, but not true parallel speedup due to the GIL.  You also have to consider the overhead of thread creation and context switching.

**28. Greedy Algorithms**

*   **Description:**  An algorithmic paradigm that makes the locally optimal choice at each step with the hope of finding a global optimum.  Greedy algorithms don't always guarantee the best solution, but they are often simple and efficient.
*   **Identification:** Problems where you can make a series of choices, and it seems intuitive to pick the "best" option at each step.  Look for keywords like "maximize," "minimize," "shortest," "largest," "most," but be careful – a greedy approach isn't always correct. You *must* be able to prove that the greedy choice property holds (that a locally optimal choice leads to a globally optimal solution).
*   **Example:** Activity Selection Problem (given a set of activities with start and finish times, select the maximum number of non-overlapping activities).


```python
def activity_selection(activities):
    # Sort activities by finish time
    activities.sort(key=lambda x: x[1])
    selected_activities = []
    last_finish_time = -1  # Initialize to a time before any activity starts

    for start_time, finish_time in activities:
        if start_time >= last_finish_time:
            selected_activities.append((start_time, finish_time))
            last_finish_time = finish_time
    return selected_activities

#example
activities = [(1, 4), (3, 5), (0, 6), (5, 7), (3, 9), (5, 9), (6, 10), (8, 11), (8, 12), (2, 14), (12, 16)]
selected = activity_selection(activities)
print(selected) # [(1, 4), (5, 7), (8, 11), (12, 16)]
```

*   **Alternatives:** Dynamic programming (often provides optimal solutions, but can be more complex), brute force (usually inefficient).
*   **Complexity:**  Often O(n log n) due to sorting, or O(n) if the input is already sorted or a suitable data structure (like a priority queue) can be used efficiently.
