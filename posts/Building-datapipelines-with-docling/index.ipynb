**II. Web Scraping with `docling`**

Web scraping – it's a classic technique for building datasets from the vast expanse of the internet. Instead of manually copying and pasting data (yuck!), we can use code to automatically extract the information we need. `docling` can make this process even smoother.

Let's say you want to grab the text content from a webpage. Here's how you can do it. **NOTE: `docling` does not provide web scraping on its own. This example uses BeautifulSoup.**

````python
# Web scraping example using BeautifulSoup
import requests
from bs4 import BeautifulSoup

url = "https://example.com"  # Replace with your target URL
response = requests.get(url)
html_content = response.text

soup = BeautifulSoup(html_content, 'html.parser') # Using BeautifulSoup to parse the HTML
text_content = soup.get_text()

# Print the extracted text
print(text_content)
````

In this example, we used the `requests` library to fetch the HTML content of a webpage, then we used BeautifulSoup to parse it and extract the text.

Want to be more specific and grab only the headings or paragraphs? You can target specific HTML elements using CSS selectors.
Remember to be a responsible scraper! Respect the `robots.txt` file of the website and avoid making too many requests in a short period of time.


````python
soup = BeautifulSoup(html_content, 'html.parser')

# Extract all paragraph elements
paragraphs = soup.find_all('p') # Find all paragraph tags
for p in paragraphs:
    print(p.text)
````

**III. Extracting Documentation from GitHub Repositories**

Now, let's move on to extracting documentation from GitHub repositories. This is super useful when you want to analyze the documentation of a library, framework, or any other open-source project.

Let's start by extracting the README.md file from a repository.

````python
import requests

repo_owner = "DS4SD"
repo_name = "docling"
readme_url = f"https://raw.githubusercontent.com/{repo_owner}/{repo_name}/master/README.md"

response = requests.get(readme_url)
readme_content = response.text

# Print the README content
print(readme_content)
````


Easy peasy! We construct the URL for the raw README file and then use `requests` to download its content.

But what if the documentation is spread across multiple files in a `docs/` directory? Here's how you can extract those as well:

````python
import requests

repo_owner = "DS4SD"
repo_name = "docling"
docs_url = f"https://api.github.com/repos/{repo_owner}/{repo_name}/contents/docs"

response = requests.get(docs_url)
docs_contents = response.json()

for doc in docs_contents:
    if doc['type'] == 'file' and doc['name'].endswith('.md'):
        doc_url = doc['download_url']
        doc_response = requests.get(doc_url)
        doc_content = doc_response.text
        print(f"Content of {doc['name']}:\n{doc_content}\n")
````

This code snippet uses the GitHub API to list the contents of the `docs/` directory. It then iterates through the files, checks if they are Markdown files, and extracts their content.

Keep in mind that the GitHub API has rate limits. If you're making a lot of requests, you might need to authenticate to increase your rate limit.

**IV. Basic Data Processing**

Okay, so now we have our data extracted!  The next step is to clean and process it so it's ready for our AI models. This usually involves:

*   Removing HTML tags or Markdown syntax
*   Tokenization (splitting text into words or sentences)
*   Lowercasing text
*   Removing stop words (common words like "the", "a", "is")
*   Stemming or lemmatization (reducing words to their root form)

Here's a code snippet that demonstrates some of these techniques:

````python
# Data Cleaning and Preprocessing
import re
import nltk
from nltk.corpus import stopwords
from nltk.tokenize import word_tokenize

nltk.download('stopwords')
nltk.download('punkt')

text = "This is an example sentence with some HTML tags <p>and punctuation!</p>"

# Remove HTML tags
text = re.sub('<[^>]*>', '', text)

# Tokenize the text
tokens = word_tokenize(text)

# Convert to lowercase
tokens = [token.lower() for token in tokens]

# Remove stop words
stop_words = set(stopwords.words('english'))
tokens = [token for token in tokens if token not in stop_words]

print(tokens)  # Output: ['example', 'sentence', 'punctuation', '!']
````

We use regular expressions to remove HTML tags, `nltk` for tokenization and stop word removal, and lowercase the tokens.

**V. Building a Simple Data Pipeline**

Let's tie it all together by creating a simple data pipeline function:

```python
# Example Data Pipeline Function

def create_data_pipeline(url=None, repo_owner=None, repo_name=None):
    """
    A simple data pipeline for extracting and preprocessing text data from either a URL or a GitHub repository.
    """
    text = ""

    if url:
        # Web scraping part
        response = requests.get(url)
        html_content = response.text
        soup = BeautifulSoup(html_content, 'html.parser')
        text = soup.get_text()
    elif repo_owner and repo_name:
        # Github documentation extraction
        readme_url = f"https://raw.githubusercontent.com/{repo_owner}/{repo_name}/master/README.md"
        response = requests.get(readme_url)
        text = response.text
    else:
        return "Please provide either a URL or GitHub repository details."

    # Data cleaning and processing part
    text = re.sub('<[^>]*>', '', text)
    tokens = word_tokenize(text)
    tokens = [token.lower() for token in tokens]
    stop_words = set(stopwords.words('english'))
    tokens = [token for token in tokens if token not in stop_words]

    return tokens

# Example usage
url = "https://example.com"
processed_data = create_data_pipeline(url=url)
print(processed_data)

repo_owner = "DS4SD"
repo_name = "docling"
processed_data = create_data_pipeline(repo_owner=repo_owner, repo_name=repo_name)
print(processed_data)
```

This function takes either a URL or GitHub repository details, extracts the text, cleans it, and returns a list of tokens.
