# Part 1: Foundations

## 1. Events & Sample Spaces: The Foundation of Everything

### 🤔 The Big Misconception

When you see **P(E) = 0.5**, what is E?

**90% of people get this wrong.** They think E is a single outcome like "getting heads" or "drawing a red card."

**This is fundamentally incorrect** and it's why probability feels so confusing.

### 💡 The Truth: Events Are Sets

**E is not an outcome—it's an EVENT, which is a SET of outcomes.**

Let me show you with an example that will rewire your thinking.

### 📖 Example: The Marble Bag

You have a bag with 10 marbles:
- **Red**: R1, R2, R3, R4, R5
- **Blue**: B1, B2, B3
- **Green**: G1, G2

````python
# Define our sample space (ALL possible outcomes)
marbles = ['R1', 'R2', 'R3', 'R4', 'R5', 'B1', 'B2', 'B3', 'G1', 'G2']

print(f"Sample Space S = {marbles}")
print(f"Total outcomes: {len(marbles)}")
````

**Sample Space (S)**: The set of ALL possible outcomes. In this case, 10 marbles.

When you reach in blindfolded, each marble has a 1/10 chance. But here's the breakthrough moment:

**Question:** "What's the probability of drawing a red marble?"

This is NOT asking about one outcome. It's asking about **five different outcomes grouped together**:

````python
# Define an EVENT: drawing a red marble
red_event = ['R1', 'R2', 'R3', 'R4', 'R5']

# An event is a COLLECTION of outcomes
print(f"Event E (Red) = {red_event}")
print(f"Number of outcomes in E: {len(red_event)}")

# Calculate probability
probability_red = len(red_event) / len(marbles)
print(f"\nP(Red) = {len(red_event)}/{len(marbles)} = {probability_red} = {probability_red*100}%")

# Output:
# Event E (Red) = ['R1', 'R2', 'R3', 'R4', 'R5']
# Number of outcomes in E: 5
# P(Red) = 5/10 = 0.5 = 50.0%
````

### 🎯 The Fundamental Formula

$$P(E) = \frac{\text{Number of outcomes in event E}}{\text{Total number of outcomes in S}}$$

### 📖 Example: Two Coin Flips

Quick: What's the sample space for flipping two coins?

**Common mistake**: {HH, HT, TT} ← Missing one!


````python
# CORRECT sample space
sample_space = ['HH', 'HT', 'TH', 'TT']

print(f"Sample space: {sample_space}")
print(f"Total outcomes: {len(sample_space)}")

# Define different events
events = {
    'Exactly one head': ['HT', 'TH'],
    'At least one head': ['HH', 'HT', 'TH'],
    'First coin is heads': ['HH', 'HT'],
    'Both coins match': ['HH', 'TT']
}

# Calculate probabilities
print("\n📊 Event Probabilities:")
for event_name, outcomes in events.items():
    prob = len(outcomes) / len(sample_space)
    print(f"P({event_name}) = {len(outcomes)}/{len(sample_space)} = {prob:.2%}")

# Output:
# Sample space: ['HH', 'HT', 'TH', 'TT']
# Total outcomes: 4
# 
# 📊 Event Probabilities:
# P(Exactly one head) = 2/4 = 50.00%
# P(At least one head) = 3/4 = 75.00%
# P(First coin is heads) = 2/4 = 50.00%
# P(Both coins match) = 2/4 = 50.00%
````

### 🎯 KEY INSIGHT

**Events are sets, not single outcomes.** Yes, you could treat a single outcome like "HH" as a singleton set with one element. But understanding that E is fundamentally a *set* is what makes everything else click.


## 2. The Three Axioms: The DNA of Probability

Every probability calculation ever made—from coin flips to quantum mechanics to AI models—follows **three fundamental laws**.

These aren't arbitrary rules. They're mathematical truths as fundamental as 2 + 2 = 4.

### 📜 The Three Axioms of Probability

#### **Axiom 1: Non-Negativity**
**P(E) ≥ 0**

Probabilities can't be negative. Makes sense—you can't have less than zero chance of something happening.

#### **Axiom 2: Certainty**
**P(S) = 1**

Something must happen. The probability of the entire sample space is 100%.

#### **Axiom 3: Addition for Disjoint Events**
**If A and B can't both happen: P(A ∪ B) = P(A) + P(B)**

If events are mutually exclusive (disjoint), just add their probabilities.

### 📖 Example: Die Roll

````python
import numpy as np

# Sample space for a die
die_outcomes = [1, 2, 3, 4, 5, 6]

# Define events
event_a = [1, 2]  # Rolling less than 3
event_b = [5]     # Rolling a 5

print("🎲 DIE ROLL EXAMPLE")
print("=" * 50)

# Axiom 1: Non-negativity
p_a = len(event_a) / len(die_outcomes)
print(f"\n✅ Axiom 1: Non-negativity")
print(f"P(A) = {p_a:.4f}")
print(f"Is P(A) ≥ 0? {p_a >= 0} ✓")

# Axiom 2: Certainty
p_all = len(die_outcomes) / len(die_outcomes)
print(f"\n✅ Axiom 2: Certainty")
print(f"P(entire sample space) = {len(die_outcomes)}/{len(die_outcomes)} = {p_all}")
print(f"Is P(S) = 1? {p_all == 1} ✓")

# Axiom 3: Addition (A and B are disjoint - can't roll both <3 AND 5)
event_a_or_b = event_a + event_b
p_b = len(event_b) / len(die_outcomes)
p_a_or_b = len(event_a_or_b) / len(die_outcomes)

print(f"\n✅ Axiom 3: Addition (Disjoint Events)")
print(f"Can you roll both <3 AND 5? No! → Disjoint events")
print(f"P(A) = {p_a:.4f}")
print(f"P(B) = {p_b:.4f}")
print(f"P(A or B) = P(A) + P(B) = {p_a:.4f} + {p_b:.4f} = {p_a + p_b:.4f}")
print(f"Verify by counting: {len(event_a_or_b)}/{len(die_outcomes)} = {p_a_or_b:.4f} ✓")

# Output:
# 🎲 DIE ROLL EXAMPLE
# ==================================================
# 
# ✅ Axiom 1: Non-negativity
# P(A) = 0.3333
# Is P(A) ≥ 0? True ✓
# 
# ✅ Axiom 2: Certainty
# P(entire sample space) = 6/6 = 1.0
# Is P(S) = 1? True ✓
# 
# ✅ Axiom 3: Addition (Disjoint Events)
# Can you roll both <3 AND 5? No! → Disjoint events
# P(A) = 0.3333
# P(B) = 0.1667
# P(A or B) = P(A) + P(B) = 0.3333 + 0.1667 = 0.5000
# Verify by counting: 3/6 = 0.5000 ✓
````

### 💡 Why These Axioms Matter

These three simple laws are the **DNA of probability**. Every complex formula you'll ever see—conditional probability, Bayes' rule, expected value—is just these axioms in disguise.

Think of it this way:
- **Axiom 1** keeps things sensible (no negative probabilities)
- **Axiom 2** ensures completeness (all probabilities sum to 1)
- **Axiom 3** gives us a way to combine events

### 🎯 The Pattern for Every Problem

From now on, every problem follows this structure:
1. **Identify the sample space** (all possible outcomes)
2. **Define your events** (sets of outcomes)
3. **Apply the axioms** (calculate probabilities)

## 3. Set Operations: Combining Events

Real life is messy. You rarely ask simple questions like "Will it rain?" More often:
- "Will it rain **OR** snow?"
- "Did I get a king **AND** a heart?"
- "Will it **NOT** rain?"

There's a beautiful mathematical structure for handling exactly these situations.

### 🎯 The Three Operations

#### **1. NOT (Complement) - The "Everything Else"**

**Notation:** A' or Ā or A^c

**Meaning:** Everything in the sample space EXCEPT A

````python
# Example: Card deck
deck_size = 52
kings = 4

# P(King)
p_king = kings / deck_size

# P(NOT King) - Everything except kings
p_not_king = 1 - p_king

print("🃏 COMPLEMENT EXAMPLE")
print("=" * 50)
print(f"P(King) = {kings}/{deck_size} = {p_king:.4f}")
print(f"P(NOT King) = 1 - {p_king:.4f} = {p_not_king:.4f}")
print(f"\nVerify: {kings} + {deck_size - kings} = {deck_size} cards ✓")
print(f"Probabilities sum to 1: {p_king:.4f} + {p_not_king:.4f} = {p_king + p_not_king:.4f} ✓")
````

**The Complement Rule:**

$$P(\text{not } A) = 1 - P(A)$$

**Why it works:** Either A happens or it doesn't. These are the only two possibilities and they must add to 1.

#### **2. AND (Intersection) - "Both Conditions"**

**Notation:** A ∩ B

**Meaning:** Outcomes that satisfy BOTH conditions

````python
# Example: King AND Heart
total_cards = 52
kings_in_deck = 4
hearts_in_deck = 13

# Which cards are BOTH king AND heart?
king_and_heart = 1  # Only the King of Hearts!

p_king_and_heart = king_and_heart / total_cards

print("🎴 INTERSECTION EXAMPLE")
print("=" * 50)
print(f"Kings in deck: {kings_in_deck}")
print(f"Hearts in deck: {hearts_in_deck}")
print(f"Cards that are BOTH king AND heart: {king_and_heart}")
print(f"\nP(King AND Heart) = {king_and_heart}/{total_cards} = {p_king_and_heart:.4f}")
````

#### **3. OR (Union) - "Either Condition" (THE TRICKY ONE)**

**Notation:** A ∪ B

**Meaning:** Outcomes that satisfy EITHER condition (or both)

**⚠️ WARNING: 80% of people mess this up!**

````python
# Example: King OR Heart

print("🎯 UNION EXAMPLE - THE MISTAKE")
print("=" * 50)

total_cards = 52
kings = 4
hearts = 13
king_of_hearts = 1  # THE OVERLAP!

# ❌ WRONG WAY (the common mistake)
wrong_answer = kings + hearts
print(f"\n❌ WRONG: Just adding")
print(f"P(King or Heart) = {kings}/{total_cards} + {hearts}/{total_cards}")
print(f"                 = {kings + hearts}/{total_cards}")
print(f"                 = {(kings + hearts)/total_cards:.4f}")
print(f"\nWhat's wrong? We counted the King of Hearts TWICE!")
print(f"Once as a king, once as a heart.")

# ✅ CORRECT WAY
correct_answer = kings + hearts - king_of_hearts
print(f"\n✅ CORRECT: Subtract the overlap")
print(f"P(King or Heart) = {kings}/{total_cards} + {hearts}/{total_cards} - {king_of_hearts}/{total_cards}")
print(f"                 = {correct_answer}/{total_cards}")
print(f"                 = {correct_answer/total_cards:.4f}")
````

### 📐 The Addition Rule (General Formula)

$$P(A \cup B) = P(A) + P(B) - P(A \cap B)$$

````python
def probability_union(p_a, p_b, p_a_and_b):
    """
    Calculate P(A or B) using the addition rule
    
    The formula automatically handles double-counting:
    - Add P(A): counts everything in A
    - Add P(B): counts everything in B (including overlap)
    - Subtract P(A and B): removes the double-count
    
    Parameters:
    -----------
    p_a : float
        Probability of event A
    p_b : float
        Probability of event B
    p_a_and_b : float
        Probability of both A and B (the overlap)
    
    Returns:
    --------
    float : P(A or B)
    """
    return p_a + p_b - p_a_and_b

# Example: Red card OR face card
p_red = 26/52     # 26 red cards
p_face = 12/52    # 12 face cards (J, Q, K in 4 suits)
p_red_and_face = 6/52  # Red face cards: J♥, Q♥, K♥, J♦, Q♦, K♦

p_red_or_face = probability_union(p_red, p_face, p_red_and_face)

print("🎴 COMPLETE EXAMPLE: Red OR Face Card")
print("=" * 50)
print(f"P(Red) = 26/52 = {p_red:.4f}")
print(f"P(Face) = 12/52 = {p_face:.4f}")
print(f"P(Red AND Face) = 6/52 = {p_red_and_face:.4f}")
print(f"\nP(Red OR Face) = {p_red:.4f} + {p_face:.4f} - {p_red_and_face:.4f}")
print(f"               = {p_red_or_face:.4f}")
print(f"               = {p_red_or_face*100:.2f}%")

# Verify by counting
red_cards = 26
face_cards = 12
red_face_cards = 6
total_red_or_face = red_cards + face_cards - red_face_cards
print(f"\n✅ Verify by counting: {red_cards} + {face_cards} - {red_face_cards} = {total_red_or_face} cards")
print(f"   That's {total_red_or_face}/52 = {total_red_or_face/52:.4f} ✓")
````

### 🎨 Visualizing with Venn Diagrams

````python
import matplotlib.pyplot as plt
from matplotlib.patches import Circle
import numpy as np

def draw_venn_diagram():
    """Visualize union, intersection, and overlap"""
    fig, axes = plt.subplots(1, 3, figsize=(15, 5))
    
    # Common setup
    for ax in axes:
        ax.set_xlim(0, 3)
        ax.set_ylim(0, 2)
        ax.set_aspect('equal')
        ax.axis('off')
    
    # Diagram 1: Just the sets
    circle1 = Circle((1, 1), 0.6, color='red', alpha=0.3, label='Event A')
    circle2 = Circle((2, 1), 0.6, color='blue', alpha=0.3, label='Event B')
    axes[0].add_patch(circle1)
    axes[0].add_patch(circle2)
    axes[0].text(0.7, 1, 'Only A', fontsize=12, ha='center', weight='bold')
    axes[0].text(2.3, 1, 'Only B', fontsize=12, ha='center', weight='bold')
    axes[0].text(1.5, 1, 'A ∩ B\n(Both)', fontsize=11, ha='center', weight='bold')
    axes[0].set_title('Venn Diagram\nShowing Overlap', fontsize=14, weight='bold')
    axes[0].legend(loc='upper right')
    
    # Diagram 2: Intersection (AND)
    circle1 = Circle((1, 1), 0.6, color='gray', alpha=0.2)
    circle2 = Circle((2, 1), 0.6, color='gray', alpha=0.2)
    axes[1].add_patch(circle1)
    axes[1].add_patch(circle2)
    
    # Highlight intersection
    theta = np.linspace(0, 2*np.pi, 100)
    x_intersect = 1.5 + 0.3*np.cos(theta)
    y_intersect = 1 + 0.3*np.sin(theta)
    axes[1].fill(x_intersect, y_intersect, color='purple', alpha=0.6)
    axes[1].text(1.5, 1, 'A AND B', fontsize=12, ha='center', weight='bold', color='white')
    axes[1].set_title('Intersection: A ∩ B\n(Both conditions)', fontsize=14, weight='bold')
    
    # Diagram 3: Union (OR)
    circle1 = Circle((1, 1), 0.6, color='red', alpha=0.5)
    circle2 = Circle((2, 1), 0.6, color='blue', alpha=0.5)
    axes[2].add_patch(circle1)
    axes[2].add_patch(circle2)
    axes[2].text(1.5, 0.3, 'A OR B\n(Either condition)', fontsize=12, ha='center', weight='bold')
    axes[2].set_title('Union: A ∪ B\n(All shaded area)', fontsize=14, weight='bold')
    
    plt.tight_layout()
    plt.show()

draw_venn_diagram()
````

### 💡 The Power of Complements

Sometimes it's easier to calculate what you DON'T want, then subtract from 1.

**Example:** Probability of at least one head in 3 coin flips

````python
# The HARD way: List all favorable outcomes
all_outcomes = ['HHH', 'HHT', 'HTH', 'HTT', 'THH', 'THT', 'TTH', 'TTT']
at_least_one_head = [outcome for outcome in all_outcomes if 'H' in outcome]
p_hard_way = len(at_least_one_head) / len(all_outcomes)

print("🪙 AT LEAST ONE HEAD IN 3 FLIPS")
print("=" * 50)
print(f"\n😓 HARD WAY: Count favorable outcomes")
print(f"Outcomes with at least 1 head: {at_least_one_head}")
print(f"That's {len(at_least_one_head)} out of {len(all_outcomes)}")
print(f"P(at least 1 head) = {p_hard_way:.4f}")

# The EASY way: Use complement
print(f"\n😎 EASY WAY: Use complement")
print(f"P(at least 1 head) = 1 - P(no heads)")
print(f"                    = 1 - P(all tails)")

p_all_tails = (1/2) ** 3  # Each flip is tails with probability 1/2
p_easy_way = 1 - p_all_tails

print(f"P(TTT) = (1/2)³ = {p_all_tails:.4f}")
print(f"P(at least 1 head) = 1 - {p_all_tails:.4f} = {p_easy_way:.4f}")

print(f"\n✅ Both methods give same answer: {p_hard_way == p_easy_way}")
print(f"💡 Complement method: 1 calculation instead of 7!")
````

### ❌ Common Mistakes to Avoid

````python
class CommonSetOperationMistakes:
    """Learn from these common errors"""
    
    @staticmethod
    def mistake_1_forgetting_overlap():
        print("❌ MISTAKE #1: Forgetting to subtract overlap")
        print("=" * 60)
        print("\nProblem: P(King OR Heart)?")
        print("\n🚫 WRONG:")
        print("   'Just add them!'")
        print("   P(King OR Heart) = 4/52 + 13/52 = 17/52")
        print("\n✅ CORRECT:")
        print("   'Account for the King of Hearts!'")
        print("   P(King OR Heart) = 4/52 + 13/52 - 1/52 = 16/52")
        print("\n💡 TIP: Always ask: 'Can these events overlap?'")
        print("-" * 60)
    
    @staticmethod
    def mistake_2_complement():
        print("\n❌ MISTAKE #2: Complement confusion")
        print("=" * 60)
        print("\nProblem: P(NOT getting a 6 on a die)?")
        print("\n🚫 WRONG:")
        print("   'It's 5/6 because there are 5 other numbers'")
        print("   (This is actually correct, but reasoning is incomplete)")
        print("\n✅ BETTER:")
        print("   P(NOT 6) = 1 - P(6) = 1 - 1/6 = 5/6")
        print("   This works for ANY event, even complex ones!")
        print("\n💡 TIP: Use the complement rule: P(not A) = 1 - P(A)")
        print("-" * 60)
    
    @staticmethod
    def show_all():
        CommonSetOperationMistakes.mistake_1_forgetting_overlap()
        CommonSetOperationMistakes.mistake_2_complement()

# Run the mistake demonstrations
CommonSetOperationMistakes.show_all()
````



## 4. Counting: When Numbers Get Astronomical

**Scenario:** You're trying to crack a 4-digit lock. How many combinations do you need to try?

Most people guess "a few hundred." 

**The answer: 10,000** (from 0000 to 9999)

That's 10 × 10 × 10 × 10 = 10⁴.

When sample spaces get huge, we need systematic counting techniques.

### 🤔 The Two Key Questions

Before ANY counting problem, ask yourself:

1. **Can I repeat?** Can I use the same thing twice?
2. **Does order matter?** Is ABC different from CBA?

These two questions determine which formula to use.

### 🌳 The Decision Tree

```
Can I repeat?
├─ YES → Does order matter?
│        ├─ YES → n^k (Repetition with order)
│        └─ NO  → Complex (rarely used)
│
└─ NO  → Does order matter?
         ├─ YES → Permutations: P(n,k) = n!/(n-k)!
         └─ NO  → Combinations: C(n,k) = n!/[k!(n-k)!]
```

### 📚 Quick Math Review: Factorials

````python
def explain_factorial(n):
    """Show what factorial means step by step"""
    result = 1
    steps = []
    
    for i in range(n, 0, -1):
        steps.append(str(i))
        result *= i
    
    print(f"{n}! = {' × '.join(steps)} = {result}")
    return result

print("📖 FACTORIAL EXAMPLES:")
print("=" * 50)
explain_factorial(4)
explain_factorial(5)
explain_factorial(6)

print(f"\n💡 Special case: 0! = 1 (by definition)")

# Output:
# 4! = 4 × 3 × 2 × 1 = 24
# 5! = 5 × 4 × 3 × 2 × 1 = 120
# 6! = 6 × 5 × 4 × 3 × 2 × 1 = 720
````


### 📋 The Four Formulas

````python
import math

class CountingFormulas:
    """All counting formulas in one place"""
    
    @staticmethod
    def with_repetition_order_matters(n, k):
        """
        Choosing k items from n, WITH repetition, order MATTERS
        Example: 4-digit PIN where digits can repeat
        
        Formula: n^k
        """
        result = n ** k
        print(f"Repetition allowed, order matters: {n}^{k} = {result:,}")
        return result
    
    @staticmethod
    def without_repetition_order_matters(n, k):
        """
        Choosing k items from n, NO repetition, order MATTERS
        Example: Top 3 finishers in a race
        
        Formula: P(n,k) = n!/(n-k)!
        Also: n × (n-1) × (n-2) × ... × (n-k+1)  [multiply down k times]
        """
        result = math.factorial(n) // math.factorial(n - k)
        print(f"No repetition, order matters (Permutations): P({n},{k}) = {result:,}")
        return result
    
    @staticmethod
    def without_repetition_order_doesnt_matter(n, k):
        """
        Choosing k items from n, NO repetition, order DOESN'T matter
        Example: Committee of 3 people from 10
        
        Formula: C(n,k) = n! / [k! × (n-k)!]
        """
        result = math.factorial(n) // (math.factorial(k) * math.factorial(n - k))
        print(f"No repetition, order doesn't matter (Combinations): C({n},{k}) = {result:,}")
        return result
    
    @staticmethod
    def permutation_intuition(n, k):
        """Show the 'multiply down' intuition for permutations"""
        print(f"\n💡 Intuition for P({n},{k}):")
        steps = []
        result = 1
        for i in range(k):
            current = n - i
            steps.append(f"{current}")
            result *= current
        print(f"   {' × '.join(steps)} = {result:,}")
        print(f"   (First choice: {n} options, second: {n-1}, ..., {k}th: {n-k+1})")
        return result

# Let's see examples of each
counting = CountingFormulas()

print("🔢 COUNTING EXAMPLES")
print("=" * 60)

print("\n1️⃣ 4-digit PIN (digits 0-9, can repeat)")
counting.with_repetition_order_matters(10, 4)

print("\n2️⃣ Password with 4 different digits")
counting.without_repetition_order_matters(10, 4)
counting.permutation_intuition(10, 4)

print("\n3️⃣ Lottery: Choose 4 numbers from 40 (order doesn't matter)")
counting.without_repetition_order_doesnt_matter(40, 4)

print("\n4️⃣ Committee of 3 from 10 people")
counting.without_repetition_order_doesnt_matter(10, 3)
````

## 5. Conditional Probability: When Information Changes Everything

**Scenario:** Weather app says 30% chance of rain. You look outside—dark clouds, wind picking up. Does that 30% still apply?

**Of course not.** The probability just changed based on what you observed.

This is **conditional probability**—the foundation of learning, diagnosis, and decision-making.

### 📖 The Marble Bag Revisited

````python

print("🎨 MARBLE BAG WITH REPLACEMENT")
print("=" * 60)

# Initial setup
marbles = {
    'red': 5,
    'blue': 3,
    'green': 2
}
total = sum(marbles.values())

print(f"Bag contains: {marbles}")
print(f"Total marbles: {total}")

# Before drawing anything
p_red_initially = marbles['red'] / total
print(f"\n📊 BEFORE DRAWING:")
print(f"P(Red) = {marbles['red']}/{total} = {p_red_initially:.4f} = {p_red_initially:.1%}")

# Now draw first marble (say it's Blue)
first_draw = 'blue'
print(f"\n🎯 DRAW #1: We draw a BLUE marble")

# Update the bag
remaining_marbles = marbles.copy()
remaining_marbles[first_draw] -= 1
remaining_total = sum(remaining_marbles.values())

print(f"\nRemaining marbles: {remaining_marbles}")
print(f"Total remaining: {remaining_total}")

# Calculate conditional probability
p_red_given_blue_drawn = remaining_marbles['red'] / remaining_total

print(f"\n📊 AFTER DRAWING BLUE:")
print(f"P(Red | Drew Blue first) = {remaining_marbles['red']}/{remaining_total} = {p_red_given_blue_drawn:.4f} = {p_red_given_blue_drawn:.1%}")

print(f"\n💡 INSIGHT:")
print(f"   Before: {p_red_initially:.1%} chance of red")
print(f"   After drawing blue: {p_red_given_blue_drawn:.1%} chance of red")
print(f"   The probability INCREASED! (More red as % of what's left)")
````

### 📐 The Conditional Probability Formula

**P(A | B) = P(A and B) / P(B)**

Read as: "The probability of A *given that* B happened"

````python

def conditional_probability(p_a_and_b, p_b):
    """
    Calculate P(A|B) = P(A and B) / P(B)
    
    Parameters:
    -----------
    p_a_and_b : float
        Probability of both A and B happening
    p_b : float
        Probability of B happening
        
    Returns:
    --------
    float : P(A|B) - probability of A given B occurred
    
    Intuition:
    ----------
    We're "zooming in" on just the cases where B happened,
    then asking: of those, how many also have A?
    """
    if p_b == 0:
        raise ValueError("Cannot condition on an impossible event (P(B) = 0)")
    
    return p_a_and_b / p_b
````

````python

```

### 🎯 The Classic Problem: Two Children

This problem breaks most people's intuition!

```python
print("👶 THE TWO-CHILD PROBLEM")
print("=" * 60)

# All possibilities for two children (order matters: first child, second child)
all_outcomes = ['BB', 'BG', 'GB', 'GG']

print("Setup: A family has two children")
print(f"Sample space: {all_outcomes}")
print("(BB = both boys, BG = first boy second girl, etc.)")

# Question 1: No information
print("\n❓ QUESTION 1: No info. What's P(both girls)?")
both_girls = ['GG']
p_both_girls_no_info = len(both_girls) / len(all_outcomes)
print(f"Answer: {len(both_girls)}/{len(all_outcomes)} = {p_both_girls_no_info:.2%}")

# Question 2: Conditional
print("\n❓ QUESTION 2: You know at least one is a girl. P(both girls)?")
print("\n🤔 Most people guess 1/2. Let's see...")

# Given at least one girl, what's possible?
at_least_one_girl = [outcome for outcome in all_outcomes if 'G' in outcome]
print(f"\nGiven 'at least one girl', possible outcomes: {at_least_one_girl}")
print(f"That rules out BB!")

# Of these, how many have both girls?
both_girls_cases = [outcome for outcome in at_least_one_girl if outcome == 'GG']
p_both_given_at_least_one = len(both_girls_cases) / len(at_least_one_girl)

print(f"\nOf those {len(at_least_one_girl)} outcomes, how many are GG? {len(both_girls_cases)}")
print(f"P(both girls | at least one girl) = {len(both_girls_cases)}/{len(at_least_one_girl)} = {p_both_given_at_least_one:.2%}")

print(f"\n🤯 SURPRISE: It's 1/3, not 1/2!")
print(f"   The condition 'at least one girl' changes the sample space!")
```

### 🎨 Visualizing Conditional Probability

```python
import matplotlib.pyplot as plt
import numpy as np

def visualize_conditional_probability():
    """Show how conditioning 'shrinks' the sample space"""
    
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))
    
    # Original sample space
    outcomes = ['BB', 'BG', 'GB', 'GG']
    colors = ['lightblue', 'lightpink', 'lightpink', 'pink']
    
    ax1.bar(outcomes, [1, 1, 1, 1], color=colors, edgecolor='black', linewidth=2)
    ax1.axhline(y=0.25, color='red', linestyle='--', linewidth=2, label='Each outcome: 25%')
    ax1.set_ylabel('Probability', fontsize=12)
    ax1.set_title('Original Sample Space\n(No information)', fontsize=14, weight='bold')
    ax1.set_ylim(0, 1.2)
    ax1.legend()
    
    # Add labels
    for i, (outcome, color) in enumerate(zip(outcomes, colors)):
        ax1.text(i, 1.05, '25%', ha='center', fontsize=11, weight='bold')
    
    # Conditional sample space
    conditional_outcomes = ['BG', 'GB', 'GG']
    conditional_colors = ['lightpink', 'lightpink', 'pink']
    
    # BB is greyed out
    ax2.bar(['BB'], [1], color='lightgray', alpha=0.3, edgecolor='black', linewidth=2)
    ax2.text(0, 0.5, 'RULED OUT', ha='center', va='center', fontsize=10, weight='bold', rotation=90)
    
    # Remaining outcomes get rescaled
    ax2.bar(['BG', 'GB', 'GG'], [1, 1, 1], color=conditional_colors, edgecolor='black', linewidth=2)
    ax2.axhline(y=1/3, color='red', linestyle='--', linewidth=2, label='Each outcome: 33.3%')
    ax2.set_ylabel('Probability', fontsize=12)
    ax2.set_title('Conditional Sample Space\n(Given: at least one girl)', fontsize=14, weight='bold')
    ax2.set_ylim(0, 1.2)
    ax2.legend()
    
    # Add labels
    for i in range(1, 4):
        ax2.text(i, 1.05, '33.3%', ha='center', fontsize=11, weight='bold')
    
    plt.tight_layout()
    plt.show()
    
    print("💡 KEY INSIGHT:")
    print("   Conditioning doesn't just filter outcomes—")
    print("   it RESCALES the probabilities among what remains!")

visualize_conditional_probability()
```

### 🎮 Modern Example: Dating App

```python
def dating_app_conditional_probability():
    """Conditional probability in dating apps"""
    
    print("💕 DATING APP EXAMPLE")
    print("=" * 60)
    
    # You swipe right on someone
    p_attractive = 0.30  # 30% of profiles you find attractive
    p_match_if_attractive = 0.05  # 5% match rate on attractive profiles
    p_match_if_not = 0.001  # 0.1% match rate on accidental swipes
    
    print("Setup:")
    print(f"  • You find {p_attractive:.0%} of profiles attractive")
    print(f"  • If attractive: {p_match_if_attractive:.0%} chance they swipe right too")
    print(f"  • If not attractive: {p_match_if_not:.1%} chance (accidental)")
    
    # Calculate P(match)
    p_match = (p_match_if_attractive * p_attractive + 
               p_match_if_not * (1 - p_attractive))
    
    print(f"\n📊 P(match) = {p_match:.2%}")
    
    # Now the interesting question: P(attractive | match)?
    # Using Bayes' rule (we'll learn this formally later)
    p_attractive_given_match = (p_match_if_attractive * p_attractive) / p_match
    
    print(f"\n🎯 KEY QUESTION: Given you matched, P(you find them attractive)?")
    print(f"   P(attractive | match) = {p_attractive_given_match:.1%}")
    
    print(f"\n💡 INSIGHT:")
    print(f"   Started with: {p_attractive:.0%} of all profiles are attractive to you")
    print(f"   After matching: {p_attractive_given_match:.0%} chance they're attractive")
    print(f"   Matching is STRONG EVIDENCE they're attractive!")

dating_app_conditional_probability()
```

### ✅ Check Your Understanding

- [ ] Can you explain P(A|B) in your own words?
- [ ] Do you understand how conditional probability "shrinks" the sample space?
- [ ] Can you identify when to use conditional probability?

```python
# 🏋️ PRACTICE PROBLEM

"""
A company's employees:
- 60% are engineers
- 40% are non-engineers
- 80% of engineers know Python
- 20% of non-engineers know Python

You meet a random employee who knows Python.
What's the probability they're an engineer?

Try solving before looking at the solution!
"""

# SOLUTION
print("🏢 COMPANY EMPLOYEE PROBLEM")
print("=" * 60)

p_engineer = 0.60
p_python_given_engineer = 0.80
p_python_given_non_engineer = 0.20

# First, find P(knows Python)
p_python = (p_python_given_engineer * p_engineer + 
            p_python_given_non_engineer * (1 - p_engineer))

print(f"P(Engineer) = {p_engineer:.0%}")
print(f"P(Python | Engineer) = {p_python_given_engineer:.0%}")
print(f"P(Python | Non-Engineer) = {p_python_given_non_engineer:.0%}")
print(f"\nP(Python) = {p_python:.0%}")

# Now use conditional probability
p_engineer_given_python = (p_python_given_engineer * p_engineer) / p_python

print(f"\n🎯 ANSWER:")
print(f"P(Engineer | knows Python) = {p_engineer_given_python:.1%}")

print(f"\n💡 INTERPRETATION:")
print(f"   If you meet someone random: {p_engineer:.0%} chance they're an engineer")
print(f"   If they know Python: {p_engineer_given_python:.0%} chance they're an engineer")
print(f"   Knowing Python is strong evidence for being an engineer!")
```

---

## 6. Independence: When Events Don't Affect Each Other

### 🤔 What Does Independence Mean?

Two events are **independent** if knowing one happened doesn't change the probability of the other.

```python
print("🎲 INDEPENDENCE INTUITION")
print("=" * 60)

# Example 1: Coin and die (INDEPENDENT)
print("\n✅ INDEPENDENT EVENTS: Coin flip and die roll")
print("   You flip a coin: Heads")
print("   Question: Does this change the probability of rolling a 6?")
print("   Answer: NO! The coin and die don't affect each other.")
print("   P(6) = 1/6, regardless of coin outcome")

# Example 2: Cards without replacement (DEPENDENT)
print("\n❌ DEPENDENT EVENTS: Drawing cards without replacement")
print("   First card: King")
print("   Question: Does this change P(2nd card is King)?")
print("   Answer: YES! Now only 3 Kings left in 51 cards.")
print("   P(2nd King | 1st King) = 3/51 ≠ 4/52")
```

### 📐 Three Ways to Test Independence

```python
def test_independence(p_a, p_b, p_a_and_b, event_a_name="A", event_b_name="B"):
    """
    Three equivalent tests for independence
    
    Events A and B are independent if ANY of these is true:
    1. P(A and B) = P(A) × P(B)
    2. P(A|B) = P(A)  [knowing B doesn't change P(A)]
    3. P(B|A) = P(B)  [knowing A doesn't change P(B)]
    """
    
    print(f"🔬 TESTING INDEPENDENCE: {event_a_name} and {event_b_name}")
    print("=" * 60)
    
    # Test 1: Joint = Product
    product = p_a * p_b
    test1_passes = abs(p_a_and_b - product) < 1e-10
    
    print(f"\n✓ TEST 1: P({event_a_name} and {event_b_name}) = P({event_a_name}) × P({event_b_name})?")
    print(f"  Left side: P({event_a_name} and {event_b_name}) = {p_a_and_b:.4f}")
    print(f"  Right side: P({event_a_name}) × P({event_b_name}) = {p_a:.4f} × {p_b:.4f} = {product:.4f}")
    print(f"  Equal? {test1_passes} {'✓' if test1_passes else '✗'}")
    
    # Test 2: Conditional equals marginal
    if p_b > 0:
        p_a_given_b = p_a_and_b / p_b
        test2_passes = abs(p_a_given_b - p_a) < 1e-10
        
        print(f"\n✓ TEST 2: P({event_a_name}|{event_b_name}) = P({event_a_name})?")
        print(f"  P({event_a_name}|{event_b_name}) = {p_a_given_b:.4f}")
        print(f"  P({event_a_name}) = {p_a:.4f}")
        print(f"  Equal? {test2_passes} {'✓' if test2_passes else '✗'}")
    
    # Final verdict
    print(f"\n{'🟢 INDEPENDENT' if test1_passes else '🔴 DEPENDENT'}")
    
    return test1_passes

# Example 1: Coin and die (independent)
print("\n📘 EXAMPLE 1: Coin flip and die roll")
p_heads = 1/2
p_six = 1/6
p_heads_and_six = 1/12

test_independence(p_heads, p_six, p_heads_and_six, "Heads", "Six")

# Example 2: Drawing cards without replacement (dependent)
print("\n\n📘 EXAMPLE 2: Two kings without replacement")
p_first_king = 4/52
p_second_king = 4/52
p_both_kings = (4/52) * (3/51)  # Actual probability

test_independence(p_first_king, p_second_king, p_both_kings, "1st King", "2nd King")
```

### 📊 Joint Probability Tables

```python
import pandas as pd

def create_joint_probability_table():
    """Visualize joint and marginal probabilities"""
    
    print("📊 JOINT PROBABILITY TABLE: Student Exam Performance")
    print("=" * 70)
    
    # Data: Out of 100 students
    data = {
        'Pass Final': [50, 5],    # [Pass mid & final, Fail mid & pass final]
        'Fail Final': [10, 35]    # [Pass mid & fail final, Fail mid & final]
    }
    index = ['Pass Midterm', 'Fail Midterm']
    
    # Create DataFrame
    joint_table = pd.DataFrame(data, index=index)
    
    print("\n📋 COUNTS (out of 100 students):")
    print(joint_table)
    
    # Add marginals
    joint_table['Total (Midterm)'] = joint_table.sum(axis=1)
    joint_table.loc['Total (Final)'] = joint_table.sum(axis=0)
    
    print("\n📋 WITH MARGINALS:")
    print(joint_table)
    
    # Convert to probabilities
    prob_table = joint_table / 100
    
    print("\n📋 PROBABILITY TABLE:")
    print(prob_table)
    
    # Test independence
    p_pass_midterm = 0.60
    p_pass_final = 0.55
    p_both_pass = 0.50
    
    expected_if_independent = p_pass_midterm * p_pass_final
    
    print(f"\n🔬 INDEPENDENCE TEST:")
    print(f"   P(Pass Midterm) = {p_pass_midterm:.2f}")
    print(f"   P(Pass Final) = {p_pass_final:.2f}")
    print(f"   P(Pass Both) observed = {p_both_pass:.2f}")
    print(f"   P(Pass Both) if independent = {expected_if_independent:.2f}")
    print(f"\n   {p_both_pass:.2f} ≠ {expected_if_independent:.2f}")
    print(f"   → DEPENDENT! Students who pass midterm tend to pass final")

create_joint_probability_table()
```

### 🎯 The Multiplication Rule

**For independent events:**

$$P(A \text{ and } B) = P(A) \times P(B)$$

**For dependent events:**

$$P(A \text{ and } B) = P(A) \times P(B|A)$$

```python
print("⚡ MULTIPLICATION RULE")
print("=" * 60)

# Independent: Flip coin twice
print("\n✅ INDEPENDENT: Flip coin twice")
p_heads_first = 1/2
p_heads_second = 1/2  # Doesn't depend on first flip
p_two_heads_independent = p_heads_first * p_heads_second
print(f"P(HH) = P(H) × P(H) = {p_heads_first} × {p_heads_second} = {p_two_heads_independent}")

# Dependent: Draw cards without replacement
print("\n❌ DEPENDENT: Draw 2 cards, both aces")
p_first_ace = 4/52
p_second_ace_given_first = 3/51  # Conditional!
p_two_aces_dependent = p_first_ace * p_second_ace_given_first
print(f"P(Both aces) = P(1st ace) × P(2nd ace | 1st ace)")
print(f"             = {p_first_ace:.4f} × {p_second_ace_given_first:.4f}")
print(f"             = {p_two_aces_dependent:.4f}")

print(f"\n💡 KEY DIFFERENCE:")
print(f"   Independent: Just multiply")
print(f"   Dependent: Use conditional probability")
```

### ⚠️ Common Misconception: Hidden Variables

```python
def hidden_variable_example():
    """Colds and jackets - seems independent but isn't!"""
    
    print("⚠️ COMMON MISTAKE: Colds and Jackets")
    print("=" * 60)
    
    # Winter observation data (10,000 people)
    print("\nWinter Data:")
    print("  40% have colds")
    print("  70% wear jackets")
    print("  30% have cold AND wear jacket")
    
    p_cold = 0.40
    p_jacket = 0.70
    p_both = 0.30
    
    expected_if_independent = p_cold * p_jacket
    
    print(f"\n🤔 Are they independent?")
    print(f"   If independent: P(both) should be {p_cold} × {p_jacket} = {expected_if_independent:.2f}")
    print(f"   Observed: P(both) = {p_both:.2f}")
    print(f"   {p_both:.2f} > {expected_if_independent:.2f} → DEPENDENT!")
    
    print(f"\n💡 WHY?")
    print(f"   Hidden variable: WINTER")
    print(f"   • Winter → more likely to catch cold")
    print(f"   • Winter → more likely to wear jacket")
    print(f"   • So cold and jacket are correlated through season")
    
    # Visualize
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
    
    # Naive thinking
    ax1.text(0.5, 0.7, 'Cold', ha='center', fontsize=16, 
             bbox=dict(boxstyle='round', facecolor='lightblue'))
    ax1.text(0.5, 0.3, 'Jacket', ha='center', fontsize=16,
             bbox=dict(boxstyle='round', facecolor='lightcoral'))
    ax1.text(0.5, 0.5, '?', ha='center', fontsize=20)
    ax1.set_xlim(0, 1)
    ax1.set_ylim(0, 1)
    ax1.axis('off')
    ax1.set_title('Naive: Are they independent?', fontsize=14, weight='bold')
    
    # Actual relationship
    ax2.text(0.5, 0.9, 'Winter', ha='center', fontsize=16,
             bbox=dict(boxstyle='round', facecolor='lightyellow'))
    ax2.arrow(0.5, 0.85, -0.15, -0.2, head_width=0.05, head_length=0.05, fc='black')
    ax2.arrow(0.5, 0.85, 0.15, -0.2, head_width=0.05, head_length=0.05, fc='black')
    ax2.text(0.25, 0.5, 'Cold', ha='center', fontsize=16,
             bbox=dict(boxstyle='round', facecolor='lightblue'))
    ax2.text(0.75, 0.5, 'Jacket', ha='center', fontsize=16,
             bbox=dict(boxstyle='round', facecolor='lightcoral'))
    ax2.set_xlim(0, 1)
    ax2.set_ylim(0, 1)
    ax2.axis('off')
    ax2.set_title('Reality: Hidden common cause', fontsize=14, weight='bold')
    
    plt.tight_layout()
    plt.show()

hidden_variable_example()
```

### ✅ Check Your Understanding

- [ ] Can you define independence in your own words?
- [ ] Do you know the three tests for independence?
- [ ] Can you explain the difference between independent and disjoint?
- [ ] Do you understand why correlation ≠ causation?

```python
# 🏋️ PRACTICE PROBLEM

"""
Smartphone notifications:
- 30% of notifications are from social media
- 20% of notifications are urgent
- 10% are both social media AND urgent

Are "social media" and "urgent" independent?
"""

# SOLUTION
print("📱 NOTIFICATION INDEPENDENCE TEST")
print("=" * 60)

p_social = 0.30
p_urgent = 0.20
p_both = 0.10

expected_if_independent = p_social * p_urgent

print(f"P(Social Media) = {p_social:.2f}")
print(f"P(Urgent) = {p_urgent:.2f}")
print(f"P(Both) observed = {p_both:.2f}")
print(f"P(Both) if independent = {expected_if_independent:.2f}")

if abs(p_both - expected_if_independent) < 0.01:
    print(f"\n✅ INDEPENDENT!")
else:
    print(f"\n❌ DEPENDENT!")
    print(f"   {p_both:.2f} > {expected_if_independent:.2f}")
    print(f"   Social media notifications are more likely to be urgent")
```

---

*Continue to Part 3: Random Variables and Expected Value...*

[Due to length constraints, I'll continue with the remaining sections. Would you like me to continue with sections 7-13, including Random Variables, Expected Value, Bayes' Rule, etc.?]
````

````python
# Part 3: Random Variables & Expected Value

## 7. Random Variables: Turning Randomness into Numbers

### 🤔 What Is a Random Variable?

Up until now, we've been dealing with outcomes like "heads," "red marble," or "King of Hearts." But what if we want to work with actual **numbers**?

A **random variable** is a function that assigns a number to every outcome in your sample space. Think of it as a translator that converts random outcomes into numbers we can calculate with.

```python
print("🎲 FROM OUTCOMES TO NUMBERS")
print("=" * 60)

# Sample space: Flip two coins
outcomes = ['HH', 'HT', 'TH', 'TT']

print("Sample Space (outcomes):", outcomes)
print("\nBut we want to work with NUMBERS...")

# Define a random variable X = number of heads
def count_heads(outcome):
    """Random variable: Count the heads"""
    return outcome.count('H')

# Apply the random variable to each outcome
random_variable_X = {}
for outcome in outcomes:
    random_variable_X[outcome] = count_heads(outcome)

print("\nRandom Variable X = 'Number of Heads':")
for outcome, value in random_variable_X.items():
    print(f"  {outcome} → X = {value}")

print("\n💡 We've translated random outcomes into numbers!")
```

### 📖 Example: The Gambling Game

```python
class GamblingGame:
    """A game where winnings depend on coin flips"""
    
    def __init__(self):
        self.outcomes = ['HH', 'HT', 'TH', 'TT']
        self.probabilities = [0.25, 0.25, 0.25, 0.25]
    
    def calculate_winnings(self, outcome):
        """
        Rules:
        - Win $5 for each head
        - Lose $2 for each tail
        """
        heads = outcome.count('H')
        tails = outcome.count('T')
        return heads * 5 - tails * 2
    
    def show_random_variable(self):
        """Display the random variable X = winnings"""
        print("💰 GAMBLING GAME")
        print("=" * 60)
        print("Rules: +$5 per head, -$2 per tail")
        print("\nRandom Variable X = 'Total Winnings':")
        
        winnings_map = {}
        for outcome in self.outcomes:
            winnings = self.calculate_winnings(outcome)
            winnings_map[outcome] = winnings
            
            heads = outcome.count('H')
            tails = outcome.count('T')
            print(f"  {outcome}: {heads}H, {tails}T → ${winnings:+d}")
        
        # Show probability distribution
        from collections import Counter
        value_counts = Counter(winnings_map.values())
        
        print("\n📊 Probability Distribution of X:")
        for value in sorted(value_counts.keys()):
            count = value_counts[value]
            prob = count / len(self.outcomes)
            print(f"  P(X = ${value:+d}) = {count}/{len(self.outcomes)} = {prob:.2%}")
        
        return winnings_map
    
    def visualize_distribution(self):
        """Plot the probability distribution"""
        winnings_map = self.show_random_variable()
        
        from collections import Counter
        value_counts = Counter(winnings_map.values())
        
        values = sorted(value_counts.keys())
        probabilities = [value_counts[v] / len(self.outcomes) for v in values]
        
        plt.figure(figsize=(10, 6))
        bars = plt.bar(values, probabilities, color='steelblue', 
                      edgecolor='black', linewidth=2, alpha=0.7)
        
        # Add value labels on bars
        for bar, prob in zip(bars, probabilities):
            height = bar.get_height()
            plt.text(bar.get_x() + bar.get_width()/2., height,
                    f'{prob:.1%}',
                    ha='center', va='bottom', fontsize=12, weight='bold')
        
        plt.xlabel('Winnings (X)', fontsize=12, weight='bold')
        plt.ylabel('Probability', fontsize=12, weight='bold')
        plt.title('Probability Distribution of Winnings', fontsize=14, weight='bold')
        plt.grid(axis='y', alpha=0.3)
        plt.axhline(y=0.25, color='red', linestyle='--', alpha=0.5, label='Uniform (0.25)')
        plt.legend()
        plt.tight_layout()
        plt.show()

# Run the game
game = GamblingGame()
game.visualize_distribution()
```

### 🎯 Random Variables Are Everywhere

```python
class RandomVariableExamples:
    """Real-world examples of random variables"""
    
    @staticmethod
    def examples():
        examples = [
            {
                'scenario': 'Roll a die',
                'random_variable': 'X = the number shown',
                'possible_values': [1, 2, 3, 4, 5, 6],
                'type': 'Discrete'
            },
            {
                'scenario': 'Measure someone\'s height',
                'random_variable': 'H = height in inches',
                'possible_values': 'Any value between ~50 and ~80',
                'type': 'Continuous'
            },
            {
                'scenario': 'Count cars at intersection',
                'random_variable': 'C = number of cars in 1 hour',
                'possible_values': [0, 1, 2, 3, ...],
                'type': 'Discrete'
            },
            {
                'scenario': 'Wait for a bus',
                'random_variable': 'T = waiting time in minutes',
                'possible_values': 'Any value ≥ 0',
                'type': 'Continuous'
            },
            {
                'scenario': 'Take a test',
                'random_variable': 'S = your score (0-100)',
                'possible_values': 'Any integer 0 to 100',
                'type': 'Discrete'
            },
            {
                'scenario': 'Stock market',
                'random_variable': 'P = price change',
                'possible_values': 'Any real number',
                'type': 'Continuous'
            }
        ]
        
        print("🌍 RANDOM VARIABLES IN REAL LIFE")
        print("=" * 70)
        
        for i, ex in enumerate(examples, 1):
            print(f"\n{i}. {ex['scenario']}")
            print(f"   Random Variable: {ex['random_variable']}")
            print(f"   Possible Values: {ex['possible_values']}")
            print(f"   Type: {ex['type']}")

RandomVariableExamples.examples()
```

### 📊 Discrete vs Continuous Random Variables

```python
print("\n📘 DISCRETE vs CONTINUOUS")
print("=" * 60)

print("\n🔢 DISCRETE Random Variable:")
print("   • Takes on COUNTABLE values (1, 2, 3, ...)")
print("   • Can list all possible values")
print("   • Examples: # of heads, # of cars, die roll")
print("   • Use PMF (Probability Mass Function)")
print("   • Can ask: P(X = exact value)")

print("\n📏 CONTINUOUS Random Variable:")
print("   • Takes on UNCOUNTABLE values (any real number in a range)")
print("   • Cannot list all values (infinitely many)")
print("   • Examples: height, weight, time, temperature")
print("   • Use PDF (Probability Density Function)")
print("   • Can only ask: P(a ≤ X ≤ b)")
```

### 🎨 Visualizing Random Variables

```python
def compare_discrete_continuous():
    """Side-by-side comparison"""
    
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 5))
    
    # Discrete: Die roll
    die_values = [1, 2, 3, 4, 5, 6]
    die_probs = [1/6] * 6
    
    bars = ax1.bar(die_values, die_probs, width=0.6, color='steelblue',
                   edgecolor='black', linewidth=2, alpha=0.7)
    
    for bar in bars:
        height = bar.get_height()
        ax1.text(bar.get_x() + bar.get_width()/2., height,
                f'{height:.3f}',
                ha='center', va='bottom', fontsize=10)
    
    ax1.set_xlabel('Value', fontsize=12, weight='bold')
    ax1.set_ylabel('Probability', fontsize=12, weight='bold')
    ax1.set_title('DISCRETE: Die Roll\n(Can ask P(X = 3))', 
                  fontsize=13, weight='bold')
    ax1.set_xticks(die_values)
    ax1.grid(axis='y', alpha=0.3)
    
    # Continuous: Uniform distribution
    x = np.linspace(0, 10, 1000)
    density = np.where((x >= 2) & (x <= 8), 1/6, 0)
    
    ax2.plot(x, density, 'b-', linewidth=3, label='PDF')
    ax2.fill_between(x, 0, density, alpha=0.3)
    
    # Shade a region to show probability
    x_region = x[(x >= 4) & (x <= 6)]
    y_region = density[(x >= 4) & (x <= 6)]
    ax2.fill_between(x_region, 0, y_region, color='red', alpha=0.5,
                     label='P(4 ≤ X ≤ 6)')
    
    ax2.set_xlabel('Value', fontsize=12, weight='bold')
    ax2.set_ylabel('Density', fontsize=12, weight='bold')
    ax2.set_title('CONTINUOUS: Uniform(2,8)\n(Can only ask P(a ≤ X ≤ b))', 
                  fontsize=13, weight='bold')
    ax2.legend()
    ax2.grid(alpha=0.3)
    ax2.set_xlim(0, 10)
    
    plt.tight_layout()
    plt.show()

compare_discrete_continuous()
```

### ✅ Check Your Understanding

- [ ] Can you define a random variable in your own words?
- [ ] Can you distinguish between discrete and continuous random variables?
- [ ] Can you give three real-world examples of each type?

```python
# 🏋️ PRACTICE PROBLEM

"""
Roll two dice. Define random variable Y = sum of the two dice.

1. List all possible values of Y
2. Calculate P(Y = 7)
3. What's the most likely sum?
"""

# SOLUTION
print("🎲🎲 TWO DICE PROBLEM")
print("=" * 60)

# Generate all outcomes
outcomes = []
for die1 in range(1, 7):
    for die2 in range(1, 7):
        outcomes.append((die1, die2))

print(f"Total outcomes: {len(outcomes)}")

# Define random variable Y = sum
sums = [die1 + die2 for die1, die2 in outcomes]

# 1. Possible values
possible_values = sorted(set(sums))
print(f"\n1. Possible values of Y: {possible_values}")

# 2. P(Y = 7)
count_seven = sums.count(7)
p_seven = count_seven / len(outcomes)
print(f"\n2. P(Y = 7) = {count_seven}/{len(outcomes)} = {p_seven:.4f} = {p_seven:.2%}")

ways_to_get_seven = [(d1, d2) for d1, d2 in outcomes if d1 + d2 == 7]
print(f"   Ways to get 7: {ways_to_get_seven}")

# 3. Most likely sum
from collections import Counter
sum_counts = Counter(sums)
most_common = sum_counts.most_common(1)[0]

print(f"\n3. Most likely sum: {most_common[0]} (appears {most_common[1]} times)")

# Visualize
plt.figure(figsize=(12, 6))
values = sorted(sum_counts.keys())
counts = [sum_counts[v] for v in values]
probs = [c / len(outcomes) for c in counts]

bars = plt.bar(values, probs, color='steelblue', edgecolor='black', linewidth=2, alpha=0.7)
bars[values.index(7)].set_color('red')  # Highlight 7

for bar, prob in zip(bars, probs):
    height = bar.get_height()
    plt.text(bar.get_x() + bar.get_width()/2., height,
            f'{prob:.3f}',
            ha='center', va='bottom', fontsize=9)

plt.xlabel('Sum (Y)', fontsize=12, weight='bold')
plt.ylabel('Probability', fontsize=12, weight='bold')
plt.title('Probability Distribution: Sum of Two Dice', fontsize=14, weight='bold')
plt.xticks(values)
plt.grid(axis='y', alpha=0.3)
plt.tight_layout()
plt.show()
```

---

## 8. Expected Value: The Long-Run Average

**Expected value** is the single most important concept in practical probability. It's the weighted average of all possible outcomes—what you expect "on average" in the long run.

### 📐 The Formula

**E[X] = Σ (value × probability of that value)**

```python
def expected_value(values, probabilities):
    """
    Calculate E[X] = Σ x · P(X = x)
    
    Parameters:
    -----------
    values : list
        Possible values the random variable can take
    probabilities : list
        Probability of each value
        
    Returns:
    --------
    float : Expected value
    
    Intuition:
    ----------
    It's a WEIGHTED average. Values that are more likely
    contribute more to the expected value.
    """
    if len(values) != len(probabilities):
        raise ValueError("Values and probabilities must have same length")
    
    if abs(sum(probabilities) - 1.0) > 1e-10:
        raise ValueError(f"Probabilities must sum to 1, got {sum(probabilities)}")
    
    return sum(v * p for v, p in zip(values, probabilities))


# Example: Simple die roll game
print("🎲 DIE ROLL GAME")
print("=" * 60)
print("Rules:")
print("  • Roll 6 → Win $12")
print("  • Roll 4 or 5 → Win $3")
print("  • Roll 1, 2, or 3 → Lose $6")

outcomes = [12, 3, -6]
probabilities = [1/6, 2/6, 3/6]

ev = expected_value(outcomes, probabilities)

print(f"\n📊 Calculation:")
print(f"E[X] = ($12)(1/6) + ($3)(2/6) + (-$6)(3/6)")
print(f"     = ${12 * 1/6:.2f} + ${3 * 2/6:.2f} + ${-6 * 3/6:.2f}")
print(f"     = ${ev:.2f}")

print(f"\n💡 This is a FAIR game!")
print(f"   Over the long run, you neither win nor lose money.")
```

### 🎮 The Law of Large Numbers

Expected value is what you get "on average" if you repeat the experiment many times.

```python
def simulate_law_of_large_numbers(num_trials=10000):
    """Demonstrate convergence to expected value"""
    
    print("📈 LAW OF LARGE NUMBERS SIMULATION")
    print("=" * 60)
    
    # Die game: 6→$12, 4-5→$3, 1-3→-$6
    def play_game():
        roll = np.random.randint(1, 7)
        if roll == 6:
            return 12
        elif roll in [4, 5]:
            return 3
        else:
            return -6
    
    # Play many times
    results = [play_game() for _ in range(num_trials)]
    cumulative_avg = np.cumsum(results) / np.arange(1, len(results) + 1)
    
    # Plot convergence
    plt.figure(figsize=(14, 6))
    
    plt.subplot(1, 2, 1)
    plt.plot(cumulative_avg, linewidth=2, alpha=0.7, color='steelblue')
    plt.axhline(y=0, color='red', linestyle='--', linewidth=2, label='Expected Value = $0')
    plt.xlabel('Number of Games Played', fontsize=12)
    plt.ylabel('Average Winnings ($)', fontsize=12)
    plt.title('Convergence to Expected Value', fontsize=14, weight='bold')
    plt.legend(fontsize=11)
    plt.grid(alpha=0.3)
    
    # Zoom in on later games
    plt.subplot(1, 2, 2)
    plt.plot(cumulative_avg[1000:], linewidth=2, alpha=0.7, color='steelblue')
    plt.axhline(y=0, color='red', linestyle='--', linewidth=2, label='Expected Value = $0')
    plt.xlabel('Number of Games Played', fontsize=12)
    plt.ylabel('Average Winnings ($)', fontsize=12)
    plt.title('Convergence (After 1000 games)', fontsize=14, weight='bold')
    plt.legend(fontsize=11)
    plt.grid(alpha=0.3)
    
    plt.tight_layout()
    plt.show()
    
    print(f"\nAfter {num_trials:,} games:")
    print(f"  Average winnings: ${np.mean(results):.4f}")
    print(f"  Expected value: $0.0000")
    print(f"  Difference: ${abs(np.mean(results) - 0):.4f}")
    print(f"\n💡 The more you play, the closer you get to E[X]!")

simulate_law_of_large_numbers(10000)
```

### 💰 Real-World Example: Insurance

```python
def insurance_decision():
    """Should you buy phone insurance?"""
    
    print("📱 PHONE INSURANCE DECISION")
    print("=" * 60)
    
    phone_cost = 800
    insurance_cost = 100
    probability_break = 0.05  # 5% chance you break it this year
    
    print(f"Phone cost: ${phone_cost}")
    print(f"Insurance cost: ${insurance_cost}/year")
    print(f"P(break phone) = {probability_break:.0%}")
    
    # Option 1: No insurance
    print(f"\n💭 OPTION 1: No Insurance")
    print(f"   Outcomes:")
    print(f"     • {1-probability_break:.0%} chance: Pay $0 (phone survives)")
    print(f"     • {probability_break:.0%} chance: Pay ${phone_cost} (phone breaks)")
    
    expected_cost_no_insurance = 0 * (1 - probability_break) + phone_cost * probability_break
    print(f"   E[Cost] = ($0)({1-probability_break}) + (${phone_cost})({probability_break})")
    print(f"           = ${expected_cost_no_insurance:.2f}")
    
    # Option 2: With insurance
    print(f"\n💭 OPTION 2: With Insurance")
    print(f"   Outcome: Always pay ${insurance_cost} (covered if breaks)")
    expected_cost_with_insurance = insurance_cost
    print(f"   E[Cost] = ${expected_cost_with_insurance:.2f}")
    
    # Decision
    print(f"\n🎯 DECISION:")
    if expected_cost_no_insurance < expected_cost_with_insurance:
        savings = expected_cost_with_insurance - expected_cost_no_insurance
        print(f"   ✅ DON'T buy insurance")
        print(f"   💰 Save ${savings:.2f} in expectation")
    else:
        print(f"   ✅ BUY insurance")
    
    print(f"\n⚠️ IMPORTANT CAVEAT:")
    print(f"   Expected value doesn't capture everything!")
    print(f"   • Can you afford to lose ${phone_cost}?")
    print(f"   • How much does that loss hurt you?")
    print(f"   • Risk aversion matters too!")

insurance_decision()
```

### 🎯 Expected Value in Decision Making

```python
class ExpectedValueDecisions:
    """Use expected value to make better decisions"""
    
    @staticmethod
    def job_offers():
        print("💼 JOB OFFER COMPARISON")
        print("=" * 60)
        
        # Job A: Stable salary
        job_a_salary = 80000
        job_a_bonus = 0
        
        # Job B: Lower base, but potential bonus
        job_b_base = 60000
        job_b_bonus_scenarios = [
            (0.30, 0),      # 30% chance: no bonus
            (0.50, 20000),  # 50% chance: $20k bonus
            (0.20, 50000),  # 20% chance: $50k bonus
        ]
        
        print(f"\n🏢 Job A: Stable")
        print(f"   Guaranteed: ${job_a_salary:,}/year")
        ev_a = job_a_salary
        print(f"   E[Salary] = ${ev_a:,}")
        
        print(f"\n🏢 Job B: Risky (base + bonus)")
        print(f"   Base: ${job_b_base:,}")
        print(f"   Bonus scenarios:")
        
        ev_bonus = 0
        for prob, bonus in job_b_bonus_scenarios:
            print(f"     • {prob:.0%} chance: +${bonus:,}")
            ev_bonus += prob * bonus
        
        ev_b = job_b_base + ev_bonus
        print(f"   E[Bonus] = ${ev_bonus:,}")
        print(f"   E[Total] = ${ev_b:,}")
        
        print(f"\n🎯 COMPARISON:")
        print(f"   Job A: ${ev_a:,}")
        print(f"   Job B: ${ev_b:,}")
        
        if ev_a > ev_b:
            print(f"   → Job A has higher expected value (+${ev_a - ev_b:,})")
        else:
            print(f"   → Job B has higher expected value (+${ev_b - ev_a:,})")
        
        print(f"\n💡 But also consider:")
        print(f"   • Risk tolerance (Job B is riskier)")
        print(f"   • Non-monetary factors (culture, growth, etc.)")
    
    @staticmethod
    def lottery():
        print("\n\n🎰 LOTTERY ANALYSIS")
        print("=" * 60)
        
        ticket_cost = 2
        jackpot = 1_000_000_000  # $1 billion
        probability_win = 1 / 300_000_000
        
        print(f"Ticket cost: ${ticket_cost}")
        print(f"Jackpot: ${jackpot:,}")
        print(f"P(win) = 1/{300_000_000:,} = {probability_win:.10f}")
        
        # Expected value
        ev_winnings = jackpot * probability_win
        ev_net = ev_winnings - ticket_cost
        
        print(f"\nE[Winnings] = ${ev_winnings:.2f}")
        print(f"E[Net] = ${ev_winnings:.2f} - ${ticket_cost} = ${ev_net:.2f}")
        
        print(f"\n🎯 VERDICT: {'Positive' if ev_net > 0 else 'Negative'} expected value")
        
        if ev_net < 0:
            print(f"   💸 On average, you lose ${abs(ev_net):.2f} per ticket")
            print(f"   ⚠️ The house always wins in the long run!")
    
    @staticmethod
    def show_all():
        ExpectedValueDecisions.job_offers()
        ExpectedValueDecisions.lottery()

ExpectedValueDecisions.show_all()
```

### 🎲 Common Distributions and Their Expected Values

```python
class CommonDistributions:
    """Expected values of common distributions"""
    
    @staticmethod
    def fair_die():
        print("🎲 FAIR DIE")
        values = list(range(1, 7))
        probs = [1/6] * 6
        ev = sum(v * p for v, p in zip(values, probs))
        print(f"   E[X] = (1+2+3+4+5+6)/6 = {ev:.1f}")
    
    @staticmethod
    def fair_coin():
        print("\n🪙 FAIR COIN (Heads=1, Tails=0)")
        ev = 1 * 0.5 + 0 * 0.5
        print(f"   E[X] = {ev:.1f}")
    
    @staticmethod
    def uniform(a, b):
        print(f"\n📊 UNIFORM({a},{b})")
        ev = (a + b) / 2
        print(f"   E[X] = ({a}+{b})/2 = {ev:.1f}")
        print(f"   (The middle value)")
    
    @staticmethod
    def show_all():
        print("📚 EXPECTED VALUES: COMMON DISTRIBUTIONS")
        print("=" * 60)
        CommonDistributions.fair_die()
        CommonDistributions.fair_coin()
        CommonDistributions.uniform(1, 10)
        CommonDistributions.uniform(0, 100)

CommonDistributions.show_all()
```

### ✅ Check Your Understanding

- [ ] Can you calculate expected value from a probability distribution?
- [ ] Do you understand what expected value represents?
- [ ] Can you use expected value to make decisions?

```python
# 🏋️ PRACTICE PROBLEM

"""
You're playing a carnival game:
• Pay $5 to play
• Roll two dice
• If sum is 7 or 11: Win $15
• If sum is 2, 3, or 12: Win $10
• Otherwise: Win nothing

Should you play? Calculate the expected value.
"""

# SOLUTION
print("🎪 CARNIVAL GAME")
print("=" * 60)

# Cost to play
cost = 5

# Generate all outcomes for two dice
outcomes = [(i, j) for i in range(1, 7) for j in range(1, 7)]
total_outcomes = len(outcomes)

# Categorize by sum
sums = [i + j for i, j in outcomes]
from collections import Counter
sum_counts = Counter(sums)

# Calculate probabilities
p_win_15 = (sum_counts[7] + sum_counts[11]) / total_outcomes
p_win_10 = (sum_counts[2] + sum_counts[3] + sum_counts[12]) / total_outcomes
p_win_0 = 1 - p_win_15 - p_win_10

print(f"Cost to play: ${cost}")
print(f"\nOutcomes:")
print(f"  • Win $15: sums 7 or 11 → P = {p_win_15:.4f}")
print(f"  • Win $10: sums 2, 3, or 12 → P = {p_win_10:.4f}")
print(f"  • Win $0: everything else → P = {p_win_0:.4f}")

# Expected winnings
ev_winnings = 15 * p_win_15 + 10 * p_win_10 + 0 * p_win_0
print(f"\nE[Winnings] = ${ev_winnings:.2f}")

# Expected profit (winnings - cost)
ev_profit = ev_winnings - cost
print(f"E[Profit] = ${ev_winnings:.2f} - ${cost} = ${ev_profit:.2f}")

if ev_profit > 0:
    print(f"\n✅ Play! Positive expected value of ${ev_profit:.2f}")
else:
    print(f"\n❌ Don't play! Negative expected value of ${ev_profit:.2f}")
    print(f"   On average, you lose ${abs(ev_profit):.2f} per game")
```

---

## 9. Linearity of Expectation: A Powerful Shortcut

Here's something **mind-blowing**: Expected values always add, **even when variables are completely dependent**.

This is one of the most useful shortcuts in all of probability.

### 🤯 The Surprising Truth

```python
print("🤯 LINEARITY OF EXPECTATION")
print("=" * 60)

# Roll a die with payouts
# 1-3: Win $4
# 4-6: Win $8

print("Setup: Roll a die")
print("  • Roll 1-3: Win $4")
print("  • Roll 4-6: Win $8")

# Define TWO random variables from the SAME roll
print("\nDefine TWO random variables:")
print("  X = Your winnings ($)")
print("  Y = The number you rolled")

# Build outcome table
outcomes_table = []
for roll in range(1, 7):
    winnings = 4 if roll <= 3 else 8
    outcomes_table.append({
        'Roll': roll,
        'X (winnings)': winnings,
        'Y (number)': roll,
        'X + Y': winnings + roll
    })

df = pd.DataFrame(outcomes_table)
print("\n📊 Outcome Table:")
print(df.to_string(index=False))

# Calculate expected values
e_x = df['X (winnings)'].mean()
e_y = df['Y (number)'].mean()
e_x_plus_y = df['X + Y'].mean()

print(f"\n📐 Expected Values:")
print(f"  E[X] = ${e_x:.2f}")
print(f"  E[Y] = {e_y:.2f}")
print(f"  E[X + Y] = {e_x_plus_y:.2f}")

print(f"\n🎯 THE MAGIC:")
print(f"  E[X] + E[Y] = ${e_x:.2f} + {e_y:.2f} = {e_x + e_y:.2f}")
print(f"  E[X + Y] = {e_x_plus_y:.2f}")
print(f"  They're EQUAL! ✓")

print(f"\n💡 Even though X and Y are COMPLETELY DEPENDENT,")
print(f"   E[X + Y] = E[X] + E[Y]")
```

### 🔍 Why Does This Work?

```python
def explain_linearity():
    """Show WHY expectations add"""
    
    print("\n🔬 WHY EXPECTATIONS ADD")
    print("=" * 60)
    
    print("\nThe secret: When we calculate E[X+Y], we can REGROUP:")
    print("\nE[X + Y] = (1/6)×(4+1) + (1/6)×(4+2) + ... + (1/6)×(8+6)")
    print("\nSeparate the X parts and Y parts:")
    print("         = (1/6)×4 + (1/6)×4 + (1/6)×4 + (1/6)×8 + (1/6)×8 + (1/6)×8")
    print("           + (1/6)×1 + (1/6)×2 + (1/6)×3 + (1/6)×4 + (1/6)×5 + (1/6)×6")
    print("\nThe X parts:")
    print("         = (1/6)[4 + 4 + 4 + 8 + 8 + 8] = E[X]")
    print("\nThe Y parts:")
    print("         = (1/6)[1 + 2 + 3 + 4 + 5 + 6] = E[Y]")
    print("\n💡 The dependencies automatically wash out when we separate and regroup!")

explain_linearity()
```

### ⚡ The Power: Birthday Problem

```python
def birthday_problem_with_linearity():
    """Use linearity to solve complex problems easily"""
    
    print("\n🎂 BIRTHDAY PROBLEM (Linearity Version)")
    print("=" * 60)
    
    n_people = 23
    n_days = 365
    
    print(f"Question: In a room of {n_people} people,")
    print(f"          what's the expected number of birthday matches?")
    
    print(f"\n😰 HARD WAY: Calculate for every possible pair...")
    print(f"   There are C({n_people},2) = {n_people*(n_people-1)//2} pairs!")
    
    print(f"\n😎 EASY WAY: Use linearity!")
    print(f"\n   For each pair of people:")
    print(f"     • P(same birthday) = 1/{n_days}")
    print(f"     • Define indicator variable: I_pair = 1 if match, 0 if not")
    print(f"     • E[I_pair] = 1 × (1/{n_days}) + 0 × ({n_days-1}/{n_days}) = 1/{n_days}")
    
    n_pairs = n_people * (n_people - 1) // 2
    expected_matches = n_pairs / n_days
    
    print(f"\n   Total pairs: {n_pairs}")
    print(f"   E[# matches] = {n_pairs} × (1/{n_days})")
    print(f"                = {expected_matches:.2f} matches")
    
    print(f"\n💡 We didn't need to consider dependencies between pairs!")
    print(f"   Linearity handles it automatically!")

birthday_problem_with_linearity()
```

### 🎯 The General Rule

```python
print("\n📘 LINEARITY OF EXPECTATION: THE RULE")
print("=" * 60)

print("""
For ANY random variables X and Y (dependent or independent):

    E[X + Y] = E[X] + E[Y]

More generally, for ANY constants a and b:

    E[aX + bY] = a·E[X] + b·E[Y]

This extends to any number of random variables:

    E[X₁ + X₂ + ... + Xₙ] = E[X₁] + E[X₂] + ... + E[Xₙ]

⚠️ IMPORTANT LIMITATION:
   This ONLY works for ADDITION!
   
   For multiplication, you need independence:
   E[X × Y] = E[X] × E[Y]  ← ONLY if X and Y are independent
""")
```

### 🏆 Real-World Application

```python
def coupon_collector_problem():
    """Classic problem solved elegantly with linearity"""
    
    print("🎟️ COUPON COLLECTOR PROBLEM")
    print("=" * 60)
    
    print("Scenario: Cereal boxes have 1 of 10 different toys.")
    print("          How many boxes do you expect to buy to collect all 10?")
    
    n = 10  # Number of different toys
    
    print(f"\n💭 INTUITION:")
    print(f"   Break into stages:")
    print(f"   • Stage 1: Get your 1st toy (any toy) → Takes 1 box")
    print(f"   • Stage 2: Get a NEW toy (9/10 are new) → Takes 10/9 boxes on avg")
    print(f"   • Stage 3: Get a NEW toy (8/10 are new) → Takes 10/8 boxes on avg")
    print(f"   • ...")
    print(f"   • Stage 10: Get the LAST toy (1/10 is new) → Takes 10/1 boxes on avg")
    
    print(f"\n📐 CALCULATION:")
    stages = []
    for i in range(1, n + 1):
        remaining = n - i + 1
        expected_boxes = n / remaining
        stages.append(expected_boxes)
        print(f"   Stage {i}: E[boxes] = {n}/{remaining} = {expected_boxes:.2f}")
    
    total_expected = sum(stages)
    
    print(f"\n🎯 LINEARITY OF EXPECTATION:")
    print(f"   E[Total boxes] = E[Stage1] + E[Stage2] + ... + E[Stage10]")
    print(f"                  = {' + '.join(f'{s:.2f}' for s in stages[:3])} + ...")
    print(f"                  = {total_expected:.2f} boxes")
    
    print(f"\n💡 Expected to buy ~{total_expected:.0f} boxes to collect all {n} toys!")
    
    # Visualize
    plt.figure(figsize=(12, 6))
    
    plt.subplot(1, 2, 1)
    plt.bar(range(1, n + 1), stages, color='steelblue', edgecolor='black', alpha=0.7)
    plt.xlabel('Stage', fontsize=12)
    plt.ylabel('Expected Boxes', fontsize=12)
    plt.title('Expected Boxes per Stage', fontsize=14, weight='bold')
    plt.grid(axis='y', alpha=0.3)
    
    plt.subplot(1, 2, 2)
    cumulative = np.cumsum(stages)
    plt.plot(range(1, n + 1), cumulative, 'o-', linewidth=2, markersize=8, color='steelblue')
    plt.axhline(y=total_expected, color='red', linestyle='--', label=f'Total = {total_expected:.1f}')
    plt.xlabel('Stage', fontsize=12)
    plt.ylabel('Cumulative Expected Boxes', fontsize=12)
    plt.title('Cumulative Progress', fontsize=14, weight='bold')
    plt.legend()
    plt.grid(alpha=0.3)
    
    plt.tight_layout()
    plt.show()

coupon_collector_problem()
```

### ✅ Check Your Understanding

- [ ] Do you understand that E[X+Y] = E[X] + E[Y] always holds?
- [ ] Can you explain why dependencies don't matter for addition?
- [ ] Do you know when the multiplication rule E[XY] = E[X]E[Y] applies?

```python
# 🏋️ PRACTICE PROBLEM

"""
You roll a die 100 times.
What's the expected sum of all 100 rolls?

Hint: Use linearity!
"""

# SOLUTION
print("🎲 100 DIE ROLLS")
print("=" * 60)

n_rolls = 100
e_single_roll = 3.5  # We know this from earlier

print(f"😰 HARD WAY:")
print(f"   Calculate probability distribution for sum of 100 dice...")
print(f"   There are 6^100 ≈ 6.5 × 10^77 possible outcomes!")
print(f"   Impossible to enumerate!")

print(f"\n😎 EASY WAY: Linearity!")
print(f"   Let X_i = result of roll i")
print(f"   E[X_i] = 3.5 for each roll")
print(f"\n   E[Sum] = E[X_1 + X_2 + ... + X_100]")
print(f"          = E[X_1] + E[X_2] + ... + E[X_100]  (linearity!)")
print(f"          = 3.5 + 3.5 + ... + 3.5  (100 times)")
print(f"          = 100 × 3.5")
print(f"          = {n_rolls * e_single_roll:.1f}")

print(f"\n✅ Expected sum: {n_rolls * e_single_roll:.0f}")
print(f"\n💡 One line of math instead of astronomical calculations!")
```

---

*Continue with sections 10-13? (Continuous Probability, Law of Total Probability, Bayes' Rule, Recursion)*
````

````python
# Part 4: Advanced Concepts

## 10. Continuous Probability: From Counts to Measurements

So far, we've dealt with things you can **count**: coin flips, die rolls, number of marbles. But what about things you **measure**? Height, weight, time, temperature—these flow continuously.

### 🤯 The Mind-Bending Truth

```python
print("🤔 THE CONTINUOUS PARADOX")
print("=" * 60)

print("\nImagine a bread slicer that cuts loaves between 14 and 18 inches.")
print("\nQuestion: What's P(exactly 16.000000... inches)?")
print("\n🤯 Answer: ZERO")

print("\nWhy?")
print("  • There are INFINITELY many possible lengths")
print("  • Any exact value with infinite precision has probability 0")
print("  • 16 is just ONE point among infinitely many")

print("\nWhat we CAN calculate:")
print("  ✅ P(15 ≤ length ≤ 17) = Some positive number")
print("  ✅ P(length < 16) = Some positive number")
print("  ❌ P(length = exactly 16.000...) = 0")

print("\n💡 For continuous variables, we ask about RANGES, not exact values!")
```

### 📊 PMF vs PDF

```python
print("\n📚 DISCRETE vs CONTINUOUS: Different Tools")
print("=" * 60)

comparison = """
┌─────────────────────┬──────────────────────┬──────────────────────┐
│                     │ DISCRETE             │ CONTINUOUS           │
├─────────────────────┼──────────────────────┼──────────────────────┤
│ Values              │ Countable            │ Uncountable          │
│                     │ (1, 2, 3, ...)       │ (any real number)    │
├─────────────────────┼──────────────────────┼──────────────────────┤
│ Examples            │ • Coin flips         │ • Height             │
│                     │ • # of cars          │ • Weight             │
│                     │ • Die rolls          │ • Time               │
│                     │ • # of customers     │ • Temperature        │
├─────────────────────┼──────────────────────┼──────────────────────┤
│ Tool                │ PMF                  │ PDF                  │
│                     │ (Probability Mass    │ (Probability Density │
│                     │  Function)           │  Function)           │
├─────────────────────┼──────────────────────┼──────────────────────┤
│ Can ask             │ P(X = exact value)   │ P(a ≤ X ≤ b)        │
│                     │ ✅ P(X = 3)          │ ✅ P(15 ≤ X ≤ 17)   │
│                     │                      │ ❌ P(X = 16)        │
├─────────────────────┼──────────────────────┼──────────────────────┤
│ How to calculate    │ Sum probabilities    │ Integrate (find area)│
│ probabilities       │ Σ P(X = x)          │ ∫ f(x) dx           │
└─────────────────────┴──────────────────────┴──────────────────────┘
"""
print(comparison)
```

### 🎨 Visualizing the Difference

```python
def visualize_discrete_vs_continuous():
    """Side-by-side comparison with clear distinctions"""
    
    fig = plt.figure(figsize=(16, 6))
    
    # Discrete: Die roll
    ax1 = plt.subplot(1, 3, 1)
    die_values = [1, 2, 3, 4, 5, 6]
    die_probs = [1/6] * 6
    
    bars = ax1.bar(die_values, die_probs, width=0.5, color='steelblue',
                   edgecolor='black', linewidth=2, alpha=0.7)
    
    # Annotate bars
    for bar, prob in zip(bars, die_probs):
        height = bar.get_height()
        ax1.text(bar.get_x() + bar.get_width()/2., height + 0.01,
                f'{prob:.3f}',
                ha='center', va='bottom', fontsize=10, weight='bold')
    
    ax1.set_xlabel('Value', fontsize=12, weight='bold')
    ax1.set_ylabel('Probability (P)', fontsize=12, weight='bold')
    ax1.set_title('PMF: Discrete Die Roll\n✅ Can ask P(X = 3)', 
                  fontsize=13, weight='bold')
    ax1.set_xticks(die_values)
    ax1.grid(axis='y', alpha=0.3)
    ax1.set_ylim(0, 0.25)
    
    # Continuous: Uniform distribution
    ax2 = plt.subplot(1, 3, 2)
    x = np.linspace(0, 10, 1000)
    density = np.where((x >= 2) & (x <= 8), 1/6, 0)
    
    ax2.plot(x, density, 'b-', linewidth=3, label='PDF')
    ax2.fill_between(x, 0, density, alpha=0.3)
    
    ax2.set_xlabel('Value', fontsize=12, weight='bold')
    ax2.set_ylabel('Density (f)', fontsize=12, weight='bold')
    ax2.set_title('PDF: Continuous Uniform(2,8)\n❌ Cannot ask P(X = 5)', 
                  fontsize=13, weight='bold')
    ax2.legend(fontsize=11)
    ax2.grid(alpha=0.3)
    ax2.set_xlim(0, 10)
    ax2.set_ylim(0, 0.25)
    
    # Show probability as area
    ax3 = plt.subplot(1, 3, 3)
    ax3.plot(x, density, 'b-', linewidth=3, label='PDF')
    ax3.fill_between(x, 0, density, alpha=0.2, color='blue')
    
    # Highlight a specific region
    x_region = x[(x >= 4) & (x <= 6)]
    y_region = density[(x >= 4) & (x <= 6)]
    ax3.fill_between(x_region, 0, y_region, color='red', alpha=0.6,
                     label='P(4 ≤ X ≤ 6)')
    
    # Calculate and show the probability
    prob_region = (6 - 4) * (1/6)
    ax3.text(5, 0.09, f'Area = {prob_region:.3f}', ha='center', fontsize=12,
            weight='bold', bbox=dict(boxstyle='round', facecolor='yellow', alpha=0.8))
    
    ax3.set_xlabel('Value', fontsize=12, weight='bold')
    ax3.set_ylabel('Density (f)', fontsize=12, weight='bold')
    ax3.set_title('Probability = AREA under curve\n✅ P(4 ≤ X ≤ 6) = shaded area', 
                  fontsize=13, weight='bold')
    ax3.legend(fontsize=11)
    ax3.grid(alpha=0.3)
    ax3.set_xlim(0, 10)
    ax3.set_ylim(0, 0.25)
    
    plt.tight_layout()
    plt.show()

visualize_discrete_vs_continuous()
```

### 📖 Example: Uniform Distribution

The simplest continuous distribution—every value in a range is equally likely.

```python
class UniformDistribution:
    """Uniform continuous distribution between a and b"""
    
    def __init__(self, a, b):
        """
        Create uniform distribution on [a, b]
        
        Properties:
        - Every point has the same density
        - PDF is a horizontal line
        - Total area under PDF = 1
        """
        if a >= b:
            raise ValueError("Must have a < b")
        
        self.a = a
        self.b = b
        self.width = b - a
        self.height = 1 / self.width  # Height to make total area = 1
    
    def pdf(self, x):
        """Probability density at point x"""
        if self.a <= x <= self.b:
            return self.height
        return 0
    
    def probability(self, x1, x2):
        """
        P(x1 ≤ X ≤ x2) - probability of a range
        
        For uniform: probability = (width of region) × (height of PDF)
        """
        # Clip to valid range
        x1 = max(x1, self.a)
        x2 = min(x2, self.b)
        
        if x1 >= x2:
            return 0
        
        width = x2 - x1
        return width * self.height
    
    def expected_value(self):
        """E[X] = midpoint of [a, b]"""
        return (self.a + self.b) / 2
    
    def visualize(self):
        """Plot the PDF"""
        x = np.linspace(self.a - 1, self.b + 1, 1000)
        y = np.array([self.pdf(xi) for xi in x])
        
        plt.figure(figsize=(12, 6))
        plt.plot(x, y, 'b-', linewidth=3, label=f'PDF: Uniform({self.a}, {self.b})')
        plt.fill_between(x, 0, y, alpha=0.3)
        
        plt.xlabel('Value', fontsize=12, weight='bold')
        plt.ylabel('Density', fontsize=12, weight='bold')
        plt.title(f'Uniform Distribution on [{self.a}, {self.b}]', 
                 fontsize=14, weight='bold')
        plt.axhline(y=self.height, color='red', linestyle='--', alpha=0.5,
                   label=f'Height = {self.height:.3f}')
        plt.legend(fontsize=11)
        plt.grid(alpha=0.3)
        plt.show()

# Example: Bread slicer
print("🍞 BREAD SLICER EXAMPLE")
print("=" * 60)

bread = UniformDistribution(14, 18)

print(f"Distribution: Uniform({bread.a}, {bread.b})")
print(f"PDF height: {bread.height:.4f}")
print(f"\nNote: Total area under PDF = {bread.width} × {bread.height} = 1 ✓")

print(f"\n❌ P(exactly 16.0000...) = 0 (single point has no area)")

print(f"\n✅ P(15 ≤ length ≤ 17)?")
p = bread.probability(15, 17)
print(f"   Area of rectangle: width × height")
print(f"                    = (17-15) × {bread.height:.4f}")
print(f"                    = {p:.4f}")
print(f"                    = {p*100:.1f}%")

print(f"\n✅ P(length < 16)?")
p_less = bread.probability(14, 16)
print(f"   P(14 ≤ length ≤ 16) = (16-14) × {bread.height:.4f} = {p_less:.4f} = {p_less*100:.1f}%")

print(f"\n📊 Expected length: {bread.expected_value():.1f} inches (the middle)")

bread.visualize()
```

### 📐 Integration = Area Under the Curve

For continuous distributions, probabilities are **areas**.

```python
print("\n🎓 INTEGRATION: FINDING AREAS")
print("=" * 60)

print("""
For continuous random variables:

    P(a ≤ X ≤ b) = ∫[from a to b] f(x) dx

Where:
- f(x) is the PDF (probability density function)
- ∫ is the integral (fancy sum that finds area)
- dx means "tiny slice of width"

Intuition:
- Chop the region into tiny rectangles
- Each rectangle has area ≈ f(x) × dx
- Sum all the tiny areas → get total probability
""")

# Simple example with uniform
print("\n📘 EXAMPLE: Uniform(0, 10)")
print("PDF: f(x) = 1/10 for 0 ≤ x ≤ 10")
print("\nP(3 ≤ X ≤ 7) = ∫[from 3 to 7] (1/10) dx")
print("             = (1/10) × (7 - 3)")
print("             = (1/10) × 4")
print("             = 0.4 = 40%")

print("\n💡 For uniform: Integration is just rectangle area!")
print("   For other distributions: Need calculus (or use Python!)")
```

### 🔔 The Normal Distribution (Bell Curve)

The most important continuous distribution in statistics.

```python
from scipy.stats import norm

class NormalDistribution:
    """The famous bell curve"""
    
    def __init__(self, mu=0, sigma=1):
        """
        Normal distribution N(μ, σ²)
        
        Parameters:
        -----------
        mu : float
            Mean (center of the bell)
        sigma : float
            Standard deviation (spread/width)
        """
        self.mu = mu
        self.sigma = sigma
        self.dist = norm(mu, sigma)
    
    def probability(self, a, b):
        """P(a ≤ X ≤ b) using integration"""
        return self.dist.cdf(b) - self.dist.cdf(a)
    
    def visualize_68_95_99_rule(self):
        """Show the famous 68-95-99.7 rule"""
        
        x = np.linspace(self.mu - 4*self.sigma, self.mu + 4*self.sigma, 1000)
        y = self.dist.pdf(x)
        
        fig, axes = plt.subplots(1, 3, figsize=(18, 5))
        
        # 68% within 1 std dev
        ax = axes[0]
        ax.plot(x, y, 'b-', linewidth=2)
        x_fill = x[(x >= self.mu - self.sigma) & (x <= self.mu + self.sigma)]
        y_fill = self.dist.pdf(x_fill)
        ax.fill_between(x_fill, 0, y_fill, alpha=0.5, color='green',
                        label=f'μ ± 1σ: ~68%')
        ax.axvline(self.mu, color='red', linestyle='--', linewidth=2, label='μ')
        ax.set_title('68% within 1 standard deviation', fontsize=13, weight='bold')
        ax.legend()
        ax.grid(alpha=0.3)
        
        # 95% within 2 std dev
        ax = axes[1]
        ax.plot(x, y, 'b-', linewidth=2)
        x_fill = x[(x >= self.mu - 2*self.sigma) & (x <= self.mu + 2*self.sigma)]
        y_fill = self.dist.pdf(x_fill)
        ax.fill_between(x_fill, 0, y_fill, alpha=0.5, color='yellow',
                        label=f'μ ± 2σ: ~95%')
        ax.axvline(self.mu, color='red', linestyle='--', linewidth=2, label='μ')
        ax.set_title('95% within 2 standard deviations', fontsize=13, weight='bold')
        ax.legend()
        ax.grid(alpha=0.3)
        
        # 99.7% within 3 std dev
        ax = axes[2]
        ax.plot(x, y, 'b-', linewidth=2)
        x_fill = x[(x >= self.mu - 3*self.sigma) & (x <= self.mu + 3*self.sigma)]
        y_fill = self.dist.pdf(x_fill)
        ax.fill_between(x_fill, 0, y_fill, alpha=0.5, color='orange',
                        label=f'μ ± 3σ: ~99.7%')
        ax.axvline(self.mu, color='red', linestyle='--', linewidth=2, label='μ')
        ax.set_title('99.7% within 3 standard deviations', fontsize=13, weight='bold')
        ax.legend()
        ax.grid(alpha=0.3)
        
        plt.tight_layout()
        plt.show()

# Standard normal
print("🔔 THE NORMAL DISTRIBUTION")
print("=" * 60)

standard_normal = NormalDistribution(mu=0, sigma=1)

print("Standard Normal: N(0, 1)")
print("  Mean μ = 0")
print("  Standard deviation σ = 1")

print("\n📊 The 68-95-99.7 Rule:")
p_1std = standard_normal.probability(-1, 1)
p_2std = standard_normal.probability(-2, 2)
p_3std = standard_normal.probability(-3, 3)

print(f"  P(-1 ≤ X ≤ 1) = {p_1std:.4f} ≈ 68%")
print(f"  P(-2 ≤ X ≤ 2) = {p_2std:.4f} ≈ 95%")
print(f"  P(-3 ≤ X ≤ 3) = {p_3std:.4f} ≈ 99.7%")

standard_normal.visualize_68_95_99_rule()

# Real-world example
print("\n\n📏 REAL-WORLD EXAMPLE: Heights")
print("=" * 60)

# Adult male heights: N(70, 3²) inches
heights = NormalDistribution(mu=70, sigma=3)

print("Adult male heights: N(70, 3²)")
print("  Mean: 70 inches (5'10\")")
print("  Std dev: 3 inches")

print("\n❓ What % of men are between 67 and 73 inches (5'7\" to 6'1\")?")
p = heights.probability(67, 73)
print(f"   P(67 ≤ height ≤ 73) = {p:.4f} = {p*100:.1f}%")

print("\n❓ What % of men are taller than 76 inches (6'4\")?")
p_tall = 1 - heights.dist.cdf(76)
print(f"   P(height > 76) = {p_tall:.4f} = {p_tall*100:.2f}%")
```

### ✅ Check Your Understanding

- [ ] Can you explain why P(X = exact value) = 0 for continuous variables?
- [ ] Do you understand that probability = area under the PDF?
- [ ] Can you work with uniform and normal distributions?

```python
# 🏋️ PRACTICE PROBLEM

"""
A bus arrives uniformly between 0 and 20 minutes from now.

1. What's the PDF?
2. P(wait 5-10 minutes)?
3. P(wait more than 15 minutes)?
4. Expected waiting time?
"""

# SOLUTION
print("🚌 BUS WAITING TIME")
print("=" * 60)

bus = UniformDistribution(0, 20)

print("1. PDF:")
print(f"   f(t) = 1/{bus.width} = {bus.height:.3f} for 0 ≤ t ≤ 20")
print(f"   f(t) = 0 otherwise")

print("\n2. P(wait 5-10 minutes)?")
p = bus.probability(5, 10)
print(f"   P(5 ≤ T ≤ 10) = (10-5) × {bus.height:.3f} = {p:.2f} = {p*100:.0f}%")

print("\n3. P(wait more than 15 minutes)?")
p_long = bus.probability(15, 20)
print(f"   P(T > 15) = P(15 ≤ T ≤ 20)")
print(f"             = (20-15) × {bus.height:.3f}")
print(f"             = {p_long:.2f} = {p_long*100:.0f}%")

print("\n4. Expected waiting time?")
ev = bus.expected_value()
print(f"   E[T] = (0 + 20)/2 = {ev:.0f} minutes")
print(f"   (The midpoint of the interval)")
```

---

## 11. Law of Total Probability: Breaking Complex Problems Apart

Sometimes a probability question seems impossible to calculate directly. The **Law of Total Probability** is your Swiss Army knife—it breaks complex problems into manageable pieces.

### 💡 The Core Idea

**Think of all the different PATHS to your outcome, calculate each path's probability, then add them up.**

```python
print("🛤️ LAW OF TOTAL PROBABILITY: THE IDEA")
print("=" * 60)

print("""
You want to get to outcome B.

But there are multiple ROUTES to get there:
  • Route 1: Through A₁, then to B
  • Route 2: Through A₂, then to B
  • Route 3: Through A₃, then to B
  • ...

Total probability of B = Sum of all route probabilities

Formally:
    P(B) = P(B|A₁)×P(A₁) + P(B|A₂)×P(A₂) + P(B|A₃)×P(A₃) + ...
    
Each term = (probability of route) × (probability of B given you're on that route)
""")
```

### 📖 The Classic Example: Two Bags

```python
class TwoBagProblem:
    """Classic probability problem solved with law of total probability"""
    
    def __init__(self):
        # Bag A: Balanced
        self.bag_a = {'red': 5, 'blue': 5}
        self.p_choose_a = 0.70
        
        # Bag B: Blue-biased
        self.bag_b = {'red': 1, 'blue': 7}
        self.p_choose_b = 0.30
    
    def visualize_setup(self):
        """Show the problem setup"""
        print("🎨 TWO-BAG MARBLE PROBLEM")
        print("=" * 60)
        print(f"\nBag A (70% chance): {self.bag_a['red']} red, {self.bag_a['blue']} blue")
        print(f"Bag B (30% chance): {self.bag_b['red']} red, {self.bag_b['blue']} blue")
        print(f"\nProcess: Randomly choose a bag, then draw one marble")
        print(f"\nQuestion: What's P(blue marble)?")
    
    def solve_step_by_step(self):
        """Solve using law of total probability"""
        
        print("\n🛤️ SOLUTION: Identify all paths to 'blue marble'")
        print("=" * 60)
        
        # Path 1: Choose A, then draw blue
        p_blue_given_a = self.bag_a['blue'] / sum(self.bag_a.values())
        path_1 = self.p_choose_a * p_blue_given_a
        
        print(f"\nPath 1: Choose Bag A → Draw Blue")
        print(f"  P(choose A) = {self.p_choose_a:.2f}")
        print(f"  P(blue | Bag A) = {self.bag_a['blue']}/{sum(self.bag_a.values())} = {p_blue_given_a:.2f}")
        print(f"  P(Path 1) = {self.p_choose_a:.2f} × {p_blue_given_a:.2f} = {path_1:.4f}")
        
        # Path 2: Choose B, then draw blue
        p_blue_given_b = self.bag_b['blue'] / sum(self.bag_b.values())
        path_2 = self.p_choose_b * p_blue_given_b
        
        print(f"\nPath 2: Choose Bag B → Draw Blue")
        print(f"  P(choose B) = {self.p_choose_b:.2f}")
        print(f"  P(blue | Bag B) = {self.bag_b['blue']}/{sum(self.bag_b.values())} = {p_blue_given_b:.4f}")
        print(f"  P(Path 2) = {self.p_choose_b:.2f} × {p_blue_given_b:.4f} = {path_2:.4f}")
        
        # Total
        total = path_1 + path_2
        
        print(f"\n🎯 TOTAL PROBABILITY:")
        print(f"  P(blue) = P(Path 1) + P(Path 2)")
        print(f"          = {path_1:.4f} + {path_2:.4f}")
        print(f"          = {total:.4f}")
        print(f"          = {total*100:.2f}%")
        
        return total
    
    def visualize_tree(self):
        """Draw probability tree"""
        
        fig, ax = plt.subplots(figsize=(14, 8))
        
        # Start node
        ax.text(0.5, 0.9, 'START', ha='center', fontsize=14, weight='bold',
                bbox=dict(boxstyle='round', facecolor='lightblue', edgecolor='black', linewidth=2))
        
        # Bag A branch
        ax.arrow(0.5, 0.88, -0.2, -0.15, head_width=0.02, head_length=0.02, 
                fc='black', ec='black', linewidth=2)
        ax.text(0.25, 0.78, '70%', ha='center', fontsize=11, style='italic')
        ax.text(0.25, 0.65, 'Bag A\n5R, 5B', ha='center', fontsize=12, weight='bold',
                bbox=dict(boxstyle='round', facecolor='lightgreen', edgecolor='black', linewidth=2))
        
        # Bag B branch
        ax.arrow(0.5, 0.88, 0.2, -0.15, head_width=0.02, head_length=0.02,
                fc='black', ec='black', linewidth=2)
        ax.text(0.75, 0.78, '30%', ha='center', fontsize=11, style='italic')
        ax.text(0.75, 0.65, 'Bag B\n1R, 7B', ha='center', fontsize=12, weight='bold',
                bbox=dict(boxstyle='round', facecolor='lightcoral', edgecolor='black', linewidth=2))
        
        # From Bag A
        ax.arrow(0.25, 0.63, -0.1, -0.15, head_width=0.02, head_length=0.02,
                fc='blue', ec='blue', linewidth=2)
        ax.text(0.12, 0.53, '50%', ha='center', fontsize=10, style='italic', color='blue')
        ax.text(0.12, 0.40, 'BLUE\n35%', ha='center', fontsize=11, weight='bold',
                bbox=dict(boxstyle='round', facecolor='lightblue', edgecolor='blue', linewidth=2))
        
        ax.arrow(0.25, 0.63, 0.1, -0.15, head_width=0.02, head_length=0.02,
                fc='red', ec='red', linewidth=2)
        ax.text(0.38, 0.53, '50%', ha='center', fontsize=10, style='italic', color='red')
        ax.text(0.38, 0.40, 'RED\n35%', ha='center', fontsize=11, weight='bold',
                bbox=dict(boxstyle='round', facecolor='lightcoral', edgecolor='red', linewidth=2))
        
        # From Bag B
        ax.arrow(0.75, 0.63, -0.1, -0.15, head_width=0.02, head_length=0.02,
                fc='blue', ec='blue', linewidth=2)
        ax.text(0.62, 0.53, '87.5%', ha='center', fontsize=10, style='italic', color='blue')
        ax.text(0.62, 0.40, 'BLUE\n26.25%', ha='center', fontsize=11, weight='bold',
                bbox=dict(boxstyle='round', facecolor='lightblue', edgecolor='blue', linewidth=2))
        
        ax.arrow(0.75, 0.63, 0.1, -0.15, head_width=0.02, head_length=0.02,
                fc='red', ec='red', linewidth=2)
        ax.text(0.88, 0.53, '12.5%', ha='center', fontsize=10, style='italic', color='red')
        ax.text(0.88, 0.40, 'RED\n3.75%', ha='center', fontsize=11, weight='bold',
                bbox=dict(boxstyle='round', facecolor='lightcoral', edgecolor='red', linewidth=2))
        
        # Final answer
        ax.text(0.5, 0.15, 'P(Blue) = 35% + 26.25% = 61.25%', ha='center', fontsize=14,
                weight='bold', bbox=dict(boxstyle='round', facecolor='yellow', 
                edgecolor='black', linewidth=3))
        
        ax.set_xlim(0, 1)
        ax.set_ylim(0, 1)
        ax.axis('off')
        ax.set_title('Probability Tree: Law of Total Probability', 
                    fontsize=16, weight='bold', pad=20)
        
        plt.tight_layout()
        plt.show()

# Run the example
problem = TwoBagProblem()
problem.visualize_setup()
problem.solve_step_by_step()
problem.visualize_tree()
```

### 🏭 Real-World Example: Quality Control

```python
def quality_control_example():
    """Manufacturing with multiple machines"""
    
    print("\n\n🏭 QUALITY CONTROL IN MANUFACTURING")
    print("=" * 60)
    
    print("\nScenario: Factory has 3 machines producing widgets")
    
    machines = [
        {'name': 'Machine 1', 'production': 0.50, 'defect_rate': 0.01},
        {'name': 'Machine 2', 'production': 0.30, 'defect_rate': 0.02},
        {'name': 'Machine 3', 'production': 0.20, 'defect_rate': 0.05},
    ]
    
    print("\n📊 Machine Details:")
    for m in machines:
        print(f"  {m['name']}: {m['production']:.0%} of production, {m['defect_rate']:.0%} defect rate")
    
    print("\n❓ Question: What's the overall defect rate?")
    
    print("\n🛤️ PATHS TO DEFECTIVE WIDGET:")
    
    total_defect_rate = 0
    for m in machines:
        path_prob = m['production'] * m['defect_rate']
        total_defect_rate += path_prob
        
        print(f"\n  Path: {m['name']} → Defective")
        print(f"    P({m['name']}) = {m['production']:.2f}")
        print(f"    P(defective | {m['name']}) = {m['defect_rate']:.2f}")
        print(f"    P(this path) = {m['production']:.2f} × {m['defect_rate']:.2f} = {path_prob:.4f}")
    
    print(f"\n🎯 TOTAL DEFECT RATE:")
    print(f"  P(defective) = {' + '.join([f'{m['production']:.2f}×{m['defect_rate']:.2f}' for m in machines])}")
    print(f"               = {total_defect_rate:.4f}")
    print(f"               = {total_defect_rate*100:.2f}%")
    
    # Visualize
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))
    
    # Production breakdown
    labels = [m['name'] for m in machines]
    sizes = [m['production'] for m in machines]
    colors = ['#ff9999', '#66b3ff', '#99ff99']
    
    ax1.pie(sizes, labels=labels, autopct='%1.0f%%', colors=colors, 
           startangle=90, textprops={'fontsize': 12, 'weight': 'bold'})
    ax1.set_title('Production Distribution', fontsize=14, weight='bold')
    
    # Defect contribution
    contributions = [m['production'] * m['defect_rate'] for m in machines]
    
    ax2.bar(labels, [c * 100 for c in contributions], color=colors, 
           edgecolor='black', linewidth=2, alpha=0.7)
    ax2.set_ylabel('Contribution to Overall Defect Rate (%)', fontsize=11, weight='bold')
    ax2.set_title('Each Machine\'s Contribution to Defects', fontsize=14, weight='bold')
    ax2.grid(axis='y', alpha=0.3)
    
    for i, (label, contrib) in enumerate(zip(labels, contributions)):
        ax2.text(i, contrib * 100 + 0.05, f'{contrib*100:.2f}%', 
                ha='center', va='bottom', fontsize=11, weight='bold')
    
    plt.tight_layout()
    plt.show()

quality_control_example()
```

### 🎯 The General Formula

```python
def law_of_total_probability(scenarios):
    """
    Calculate P(B) when B can happen through multiple scenarios
    
    Parameters:
    -----------
    scenarios : list of tuples
        Each tuple is (P(scenario), P(B|scenario))
        
    Returns:
    --------
    float : P(B) = Σ P(B|Aᵢ) × P(Aᵢ)
    
    Intuition:
    ----------
    Sum over all possible ways that B can happen.
    Each way contributes: (prob of being in that scenario) × (prob of B in that scenario)
    """
    
    total = sum(p_scenario * p_b_given_scenario 
                for p_scenario, p_b_given_scenario in scenarios)
    
    # Verify scenarios partition the space
    total_scenario_prob = sum(p_scenario for p_scenario, _ in scenarios)
    if abs(total_scenario_prob - 1.0) > 1e-10:
        print(f"⚠️ Warning: Scenarios don't partition space (sum to {total_scenario_prob})")
    
    return total


print("📐 LAW OF TOTAL PROBABILITY: FORMULA")
print("=" * 60)

print("""
Given:
• Event B (what we want the probability of)
• Events A₁, A₂, ..., Aₙ that partition the sample space
  (mutually exclusive and exhaustive - exactly one must happen)

Then:
    P(B) = P(B|A₁)×P(A₁) + P(B|A₂)×P(A₂) + ... + P(B|Aₙ)×P(Aₙ)
    
Or more compactly:
    P(B) = Σ P(B|Aᵢ) × P(Aᵢ)

Think of it as:
    Total probability = Sum of all pathway probabilities
""")
```

### ✅ Check Your Understanding

- [ ] Can you identify when to use the law of total probability?
- [ ] Can you enumerate all paths to an outcome?
- [ ] Do you understand how to calculate each path's probability?

```python
# 🏋️ PRACTICE PROBLEM

"""
A company uses 3 email providers:
• Gmail: 60% of employees, 98% uptime
• Outlook: 30% of employees, 95% uptime  
• Yahoo: 10% of employees, 90% uptime

An employee tries to send an email.
What's the probability it goes through successfully?
"""

# SOLUTION
print("📧 EMAIL DELIVERY PROBLEM")
print("=" * 60)

providers = [
    ('Gmail', 0.60, 0.98),
    ('Outlook', 0.30, 0.95),
    ('Yahoo', 0.10, 0.90),
]

print("Setup:")
for name, usage, uptime in providers:
    print(f"  {name}: {usage:.0%} usage, {uptime:.0%} uptime")

print("\n🛤️ Paths to successful delivery:")

scenarios = []
for name, usage, uptime in providers:
    print(f"\n  Path: Employee uses {name} → Email delivers")
    print(f"    P({name}) = {usage:.2f}")
    print(f"    P(delivers | {name}) = {uptime:.2f}")
    print(f"    P(this path) = {usage:.2f} × {uptime:.2f} = {usage * uptime:.4f}")
    scenarios.append((usage, uptime))

p_success = law_of_total_probability(scenarios)

print(f"\n🎯 ANSWER:")
print(f"  P(email delivers) = {p_success:.4f} = {p_success*100:.2f}%")
```

---

## 12. Bayes' Rule: The Crown Jewel of Probability

**Bayes' Rule** is the mathematical foundation of:
- Learning and updating beliefs
- Medical diagnosis
- Spam filters
- AI and machine learning
- Rational thinking under uncertainty

It tells us **exactly how to update our beliefs when we get new evidence**.

### 📜 The Story Behind Bayes

```python
print("📜 THE REVEREND THOMAS BAYES (1701-1761)")
print("=" * 60)

print("""
Thomas Bayes was an English minister who never published his most
famous work during his lifetime. After his death, his friend Richard
Price found the manuscript and recognized its revolutionary importance.

THE REVOLUTIONARY IDEA:
-----------------------
Before Bayes:
  Probability was about FREQUENCIES
  "I flipped a coin 100 times, got 55 heads"
  → P(heads) = 55/100 = 0.55

After Bayes:
  Probability became about BELIEF UPDATING
  "I think this coin is fair (prior = 0.5)"
  "I saw 55 heads in 100 flips (evidence)"
  "What should I believe now? (posterior = ?)"

This transformed probability from a counting exercise
into a tool for rational learning!

MODERN IMPACT:
--------------
🔹 Spam Filters: Your email learns what's spam
🔹 Medical Diagnosis: Interpreting test results correctly
🔹 Self-Driving Cars: Updating beliefs about surroundings
🔹 Machine Learning: Most AI is Bayesian belief updating
🔹 Science: How we update theories with new data
""")
```

### 🤯 The Shocking Medical Test Example

This example breaks most people's intuition!

```python
def medical_test_paradox():
    """The counterintuitive medical test result"""
    
    print("\n🏥 THE MEDICAL TEST PARADOX")
    print("=" * 60)
    
    print("\nScenario:")
    print("  • A disease affects 1% of the population")
    print("  • There's a test that's 99% accurate:")
    print("    - If you have the disease: 99% chance test is positive")
    print("    - If you don't have it: 99% chance test is negative")
    print("\n  You test POSITIVE. 😰")
    print("\n  Question: What's the probability you actually have the disease?")
    
    print("\n🤔 Most people guess: ~99%")
    print("   'The test is 99% accurate, so I'm 99% likely to have it!'")
    
    print("\n❌ WRONG!")
    
    print("\n✅ Let's calculate with 10,000 people:")
    
    # Simulate population
    population = 10_000
    disease_rate = 0.01
    test_accuracy = 0.99
    
    has_disease = int(population * disease_rate)
    no_disease = population - has_disease
    
    print(f"\n  Population: {population:,}")
    print(f"  Have disease: {has_disease} (1%)")
    print(f"  Don't have disease: {no_disease:,} (99%)")
    
    # Test results for people WITH disease
    true_positives = int(has_disease * test_accuracy)
    false_negatives = has_disease - true_positives
    
    # Test results for people WITHOUT disease
    false_positives = int(no_disease * (1 - test_accuracy))
    true_negatives = no_disease - false_positives
    
    print(f"\n  📊 TEST RESULTS:")
    print(f"     People WITH disease who test positive: {true_positives} (true positives)")
    print(f"     People WITHOUT disease who test positive: {false_positives} (FALSE positives!)")
    print(f"     ")
    print(f"     Total who test positive: {true_positives + false_positives}")
    
    # Calculate probability
    p_disease_given_positive = true_positives / (true_positives + false_positives)
    
    print(f"\n  🎯 ANSWER:")
    print(f"     P(disease | positive test) = {true_positives}/{true_positives + false_positives}")
    print(f"                                 = {p_disease_given_positive:.4f}")
    print(f"                                 = {p_disease_given_positive*100:.1f}%")
    
    print(f"\n  🤯 Only {p_disease_given_positive*100:.0f}% chance, even with a positive test!")
    
    print(f"\n  💡 WHY SO LOW?")
    print(f"     The disease is RARE (1%), so:")
    print(f"     • Only {has_disease} people actually have it")
    print(f"     • But {false_positives} healthy people test positive (1% of 9,900)")
    print(f"     • False positives OVERWHELM true positives!")
    
    # Visualize
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(16, 6))
    
    # Population breakdown
    categories = ['Have Disease\n(True Pos)', 'Have Disease\n(False Neg)', 
                  'Healthy\n(False Pos)', 'Healthy\n(True Neg)']
    counts = [true_positives, false_negatives, false_positives, true_negatives]
    colors = ['red', 'pink', 'orange', 'lightgreen']
    
    bars = ax1.bar(categories, counts, color=colors, edgecolor='black', linewidth=2, alpha=0.7)
    ax1.set_ylabel('Number of People', fontsize=12, weight='bold')
    ax1.set_title('Test Results Breakdown (Population: 10,000)', fontsize=14, weight='bold')
    ax1.grid(axis='y', alpha=0.3)
    
    for bar, count in zip(bars, counts):
        height = bar.get_height()
        ax1.text(bar.get_x() + bar.get_width()/2., height + 50,
                f'{count}',
                ha='center', va='bottom', fontsize=11, weight='bold')
    
    # Test positive breakdown
    ax2.bar(['True\nPositive', 'False\nPositive'], 
           [true_positives, false_positives],
           color=['red', 'orange'], edgecolor='black', linewidth=2, alpha=0.7)
    ax2.set_ylabel('Number of People', fontsize=12, weight='bold')
    ax2.set_title('Among Those Who Test Positive', fontsize=14, weight='bold')
    ax2.grid(axis='y', alpha=0.3)
    
    ax2.text(0, true_positives + 2, f'{true_positives}\n({p_disease_given_positive:.0%})',
            ha='center', va='bottom', fontsize=12, weight='bold')
    ax2.text(1, false_positives + 2, f'{false_positives}\n({1-p_disease_given_positive:.0%})',
            ha='center', va='bottom', fontsize=12, weight='bold')
    
    plt.tight_layout()
    plt.show()
    
    return p_disease_given_positive

medical_test_paradox()
```

### 📐 Bayes' Rule: The Formula

```python
print("\n\n📐 BAYES' RULE: THE FORMULA")
print("=" * 60)

print("""
P(A|B) = [P(B|A) × P(A)] / P(B)

Where:
------
P(A|B) = POSTERIOR
         What we want: probability of A given we observed B
         "Updated belief after seeing evidence"

P(B|A) = LIKELIHOOD
         How likely is the evidence if A is true?
         "How diagnostic is this evidence?"

P(A) = PRIOR
       Our initial belief before seeing evidence
       "What did we think before?"

P(B) = EVIDENCE
       Total probability of observing B
       "How common is this evidence overall?"
       Calculate using Law of Total Probability

THE THREE-STEP PROCESS:
-----------------------
1. Start with PRIOR belief: P(A)
2. See some EVIDENCE: B
3. Update to POSTERIOR belief: P(A|B)

The formula tells you EXACTLY how much to update!
""")
```

### 🎨 Visualizing Bayesian Updating

```python
def visualize_bayesian_updating():
    """Show how beliefs update with evidence"""
    
    # Medical test example
    prior = 0.01  # 1% disease rate
    likelihood = 0.99  # Test sensitivity
    false_positive_rate = 0.01  # 1% false positive
    
    # Calculate evidence: P(positive test)
    evidence = likelihood * prior + false_positive_rate * (1 - prior)
    
    # Calculate posterior
    posterior = (likelihood * prior) / evidence
    
    fig, axes = plt.subplots(1, 3, figsize=(18, 5))
    
    # Prior
    ax = axes[0]
    ax.bar(['Has Disease', 'Healthy'], [prior, 1-prior], 
          color=['red', 'green'], edgecolor='black', linewidth=2, alpha=0.7)
    ax.set_ylim(0, 1)
    ax.set_ylabel('Probability', fontsize=12, weight='bold')
    ax.set_title('PRIOR\n(Before test)', fontsize=14, weight='bold')
    ax.text(0, prior + 0.05, f'{prior:.0%}', ha='center', fontsize=14, weight='bold')
    ax.text(1, 0.55, f'{1-prior:.0%}', ha='center', fontsize=14, weight='bold')
    ax.grid(axis='y', alpha=0.3)
    
    # Evidence
    ax = axes[1]
    ax.text(0.5, 0.6, 'EVIDENCE', ha='center', fontsize=18, weight='bold')
    ax.text(0.5, 0.45, 'Test Result:', ha='center', fontsize=14)
    ax.text(0.5, 0.35, 'POSITIVE ⚠️', ha='center', fontsize=16, weight='bold', color='red',
           bbox=dict(boxstyle='round', facecolor='yellow', edgecolor='red', linewidth=3))
    ax.text(0.5, 0.2, f'P(positive) = {evidence:.2%}', ha='center', fontsize=12, style='italic')
    ax.set_xlim(0, 1)
    ax.set_ylim(0, 1)
    ax.axis('off')
    ax.set_title('⚡ NEW INFORMATION\n(Observe evidence)', fontsize=14, weight='bold')
    
    # Posterior
    ax = axes[2]
    ax.bar(['Has Disease', 'Healthy'], [posterior, 1-posterior],
          color=['red', 'green'], edgecolor='black', linewidth=2, alpha=0.7)
    ax.set_ylim(0, 1)
    ax.set_ylabel('Probability', fontsize=12, weight='bold')
    ax.set_title('POSTERIOR\n(After test)', fontsize=14, weight='bold')
    ax.text(0, posterior + 0.05, f'{posterior:.0%}', ha='center', fontsize=14, weight='bold')
    ax.text(1, 0.55, f'{1-posterior:.0%}', ha='center', fontsize=14, weight='bold')
    ax.grid(axis='y', alpha=0.3)
    
    # Add arrow annotations
    fig.text(0.32, 0.05, '➡️', fontsize=40, ha='center')
    fig.text(0.68, 0.05, '➡️', fontsize=40, ha='center')
    
    plt.tight_layout()
    plt.show()
    
    print(f"\n📊 BAYESIAN UPDATE SUMMARY:")
    print(f"   Prior: {prior:.1%} → Posterior: {posterior:.1%}")
    print(f"   Belief increased by factor of {posterior/prior:.1f}x")
    print(f"   But still only {posterior:.0%} likely!")

visualize_bayesian_updating()
```

### 🎯 Back to the Two-Bag Problem

```python
def two_bag_bayesian():
    """Use Bayes' rule on the marble problem"""
    
    print("\n🎨 TWO-BAG PROBLEM: BAYESIAN FLIP")
    print("=" * 60)
    
    print("\nRecall:")
    print("  Bag A (70%): 5 red, 5 blue")
    print("  Bag B (30%): 1 red, 7 blue")
    
    print("\n  We drew a BLUE marble.")
    print("\n  Question: What's P(it came from Bag A)?")
    
    # Known values
    prior_a = 0.70  # P(Bag A)
    likelihood_blue_given_a = 0.50  # P(blue | Bag A)
    evidence_blue = 0.6125  # P(blue) - calculated earlier
    
    print(f"\n📊 GIVEN INFORMATION:")
    print(f"   Prior: P(Bag A) = {prior_a:.2f}")
    print(f"   Likelihood: P(blue | Bag A) = {likelihood_blue_given_a:.2f}")
    print(f"   Evidence: P(blue) = {evidence_blue:.4f}")
    
    # Apply Bayes' rule
    posterior_a = (likelihood_blue_given_a * prior_a) / evidence_blue
    
    print(f"\n📐 BAYES' RULE:")
    print(f"   P(Bag A | blue) = P(blue | Bag A) × P(Bag A) / P(blue)")
    print(f"                   = {likelihood_blue_given_a:.2f} × {prior_a:.2f} / {evidence_blue:.4f}")
    print(f"                   = {likelihood_blue_given_a * prior_a:.4f} / {evidence_blue:.4f}")
    print(f"                   = {posterior_a:.4f}")
    print(f"                   = {posterior_a*100:.1f}%")
    
    print(f"\n💡 INTERPRETATION:")
    print(f"   Before drawing: {prior_a:.0%} confidence it was Bag A")
    print(f"   After seeing blue: {posterior_a:.1%} confidence it was Bag A")
    print(f"   ")
    print(f"   Our confidence DROPPED!")
    print(f"   Why? Blue marble is STRONGER evidence for Bag B")
    print(f"   (Bag B has more blues: 7/8 vs 5/10)")
    
    # Visualize
    fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))
    
    # Prior
    ax1.bar(['Bag A', 'Bag B'], [prior_a, 1-prior_a],
           color=['lightgreen', 'lightcoral'], edgecolor='black', linewidth=2, alpha=0.7)
    ax1.set_ylim(0, 1)
    ax1.set_ylabel('Probability', fontsize=12, weight='bold')
    ax1.set_title('BEFORE Drawing\n(Prior Belief)', fontsize=14, weight='bold')
    ax1.text(0, prior_a + 0.05, f'{prior_a:.0%}', ha='center', fontsize=16, weight='bold')
    ax1.text(1, 0.35, f'{1-prior_a:.0%}', ha='center', fontsize=16, weight='bold')
    ax1.grid(axis='y', alpha=0.3)
    
    # Posterior
    ax2.bar(['Bag A', 'Bag B'], [posterior_a, 1-posterior_a],
           color=['lightgreen', 'lightcoral'], edgecolor='black', linewidth=2, alpha=0.7)
    ax2.set_ylim(0, 1)
    ax2.set_ylabel('Probability', fontsize=12, weight='bold')
    ax2.set_title('AFTER Drawing Blue\n(Posterior Belief)', fontsize=14, weight='bold')
    ax2.text(0, posterior_a + 0.05, f'{posterior_a:.1%}', ha='center', fontsize=16, weight='bold')
    ax2.text(1, 0.48, f'{1-posterior_a:.1%}', ha='center', fontsize=16, weight='bold')
    ax2.grid(axis='y', alpha=0.3)
    
    # Add annotation
    ax2.annotate('', xy=(0, 0.6), xytext=(0, 0.7),
                arrowprops=dict(arrowstyle='->', lw=3, color='red'))
    ax2.text(0.15, 0.65, 'Decreased!', fontsize=11, weight='bold', color='red')
    
    plt.tight_layout()
    plt.show()

two_bag_bayesian()
```

### 🧠 Bayesian Learning Framework

```python
class BayesianLearner:
    """Framework for Bayesian reasoning"""
    
    def __init__(self, prior, likelihood, evidence):
        """
        Create a Bayesian reasoner
        
        Parameters:
        -----------
        prior : float
            P(hypothesis) - initial belief
        likelihood : float
            P(evidence | hypothesis) - how likely evidence if hypothesis true
        evidence : float
            P(evidence) - total probability of evidence
        """
        self.prior = prior
        self.likelihood = likelihood
        self.evidence = evidence
        self.posterior = None
    
    def update(self):
        """Apply Bayes' rule to get posterior"""
        self.posterior = (self.likelihood * self.prior) / self.evidence
        return self.posterior
    
    def summary(self):
        """Display the Bayesian update"""
        if self.posterior is None:
            self.update()
        
        print("🧠 BAYESIAN UPDATE SUMMARY")
        print("=" * 60)
        print(f"  Prior belief:     P(H) = {self.prior:.4f}")
        print(f"  Likelihood:       P(E|H) = {self.likelihood:.4f}")
        print(f"  Evidence:         P(E) = {self.evidence:.4f}")
        print(f"  Posterior belief: P(H|E) = {self.posterior:.4f}")
        
        change = self.posterior - self.prior
        direction = "increased" if change > 0 else "decreased"
        print(f"\n  💡 Belief {direction} by {abs(change):.4f}")
        print(f"     Ratio: {self.posterior/self.prior:.2f}x")

# Example: Spam filter
print("\n📧 SPAM FILTER EXAMPLE")
print("=" * 60)

print("\nYou receive an email containing the word 'FREE'")
print("Is it spam?")

print("\n📊 Known information:")
print("  • 30% of all emails are spam")
print("  • 90% of spam contains 'FREE'")
print("  • 20% of legitimate emails contain 'FREE'")

prior_spam = 0.30
p_free_given_spam = 0.90
p_free_given_legit = 0.20

# Calculate evidence: P(FREE)
evidence_free = p_free_given_spam * prior_spam + p_free_given_legit * (1 - prior_spam)

spam_filter = BayesianLearner(
    prior=prior_spam,
    likelihood=p_free_given_spam,
    evidence=evidence_free
)

posterior_spam = spam_filter.update()
spam_filter.summary()

print(f"\n🎯 DECISION:")
if posterior_spam > 0.5:
    print(f"   Classify as SPAM ({posterior_spam:.1%} confidence)")
else:
    print(f"   Classify as LEGITIMATE ({1-posterior_spam:.1%} confidence)")
```

### ✅ Check Your Understanding

- [ ] Can you explain Bayes' rule in your own words?
- [ ] Do you understand the difference between prior, likelihood, and posterior?
- [ ] Can you apply Bayes' rule to update beliefs?

```python
# 🏋️ PRACTICE PROBLEM

"""
A factory has two machines:
• Machine A: Makes 60% of products, 5% defect rate
• Machine B: Makes 40% of products, 10% defect rate

You find a defective product.
What's the probability it came from Machine A?

Try solving before looking at the solution!
"""

# SOLUTION
print("\n\n🏭 DEFECTIVE PRODUCT PROBLEM")
print("=" * 60)

# Given
p_a = 0.60
p_b = 0.40
p_defect_given_a = 0.05
p_defect_given_b = 0.10

print("Given:")
print(f"  P(Machine A) = {p_a:.0%}")
print(f"  P(Machine B) = {p_b:.0%}")
print(f"  P(defect | A) = {p_defect_given_a:.0%}")
print(f"  P(defect | B) = {p_defect_given_b:.0%}")

# Calculate evidence using law of total probability
p_defect = p_defect_given_a * p_a + p_defect_given_b * p_b

print(f"\nStep 1: Calculate P(defect)")
print(f"  P(defect) = P(defect|A)×P(A) + P(defect|B)×P(B)")
print(f"            = {p_defect_given_a}×{p_a} + {p_defect_given_b}×{p_b}")
print(f"            = {p_defect:.4f}")

# Apply Bayes' rule
p_a_given_defect = (p_defect_given_a * p_a) / p_defect

print(f"\nStep 2: Apply Bayes' Rule")
print(f"  P(A | defect) = P(defect|A) × P(A) / P(defect)")
print(f"                = {p_defect_given_a} × {p_a} / {p_defect:.4f}")
print(f"                = {p_a_given_defect:.4f}")
print(f"                = {p_a_given_defect*100:.1f}%")

print(f"\n💡 INTERPRETATION:")
print(f"   Before: {p_a:.0%} of products from Machine A")
print(f"   After finding defect: {p_a_given_defect:.1%} from Machine A")
print(f"   ")
print(f"   Probability DECREASED because Machine B has higher defect rate!")
```

---

## 13. Recursion: When the Answer Depends on Itself

Sometimes probability problems have an elegant **recursive structure**—where the answer depends on the answer itself. Instead of enumerating infinite possibilities, we turn them into simple algebra.

### 🤔 The Motivating Problem

```python
print("🪙 THE MOTIVATING PROBLEM")
print("=" * 60)

print("""
Question: How many coin flips do you expect before getting your first heads?

🤯 THE CHALLENGE:
  • You could get heads on flip 1
  • Or tails then heads (2 flips)
  • Or tails, tails, then heads (3 flips)
  • Or tails forever...?
  
  The sample space is INFINITE!
  
  Outcomes: H, TH, TTH, TTTH, TTTTH, ...
  
  How do we add up infinitely many possibilities?
""")
```

### 💡 The Recursive Insight

```python
def first_heads_recursive():
    """Solve using recursion instead of enumeration"""
    
    print("\n🧠 RECURSIVE SOLUTION")
    print("=" * 60)
    
    print("\nKey Idea: Define the answer in terms of ITSELF")
    print("\nLet E = expected number of flips to get first heads")
    
    print("\n🎯 What happens on the FIRST flip?")
    print("   • You flip once (that's 1 flip for sure)")
    print("   • Then two possibilities:")
    print("     - Heads (prob 1/2): Done! 0 more flips needed")
    print("     - Tails (prob 1/2): Back to square one, need E more flips")
    
    print("\n📐 RECURSIVE EQUATION:")
    print("   E = 1 + (1/2)×0 + (1/2)×E")
    print("       ↑     ↑         ↑")
    print("       |     |         └─ If tails: back to start")
    print("       |     └─ If heads: done")
    print("       └─ First flip always costs 1")
    
    print("\n🔧 SOLVE FOR E:")
    print("   E = 1 + 0 + E/2")
    print("   E = 1 + E/2")
    print("   E - E/2 = 1")
    print("   E/2 = 1")
    print("   E = 2")
    
    print("\n✅ ANSWER: Expected 2 flips to get first heads")
    
    # Verify with simulation
    import random
    
    def simulate_flips_to_heads(trials=100000):
        results = []
        for _ in range(trials):
            flips = 0
            while True:
                flips += 1
                if random.choice(['H', 'T']) == 'H':
                    break
            results.append(flips)
        return results
    
    print("\n🎲 SIMULATION (100,000 trials):")
    results = simulate_flips_to_heads(100000)
    sim_mean = np.mean(results)
    print(f"   Average flips: {sim_mean:.4f}")
    print(f"   Theoretical: 2.0000")
    print(f"   Difference: {abs(sim_mean - 2):.4f} ✓")
    
    # Visualize distribution
    plt.figure(figsize=(12, 6))
    
    plt.subplot(1, 2, 1)
    counts = pd.Series(results).value_counts().sort_index()
    plt.bar(counts.index[:15], counts.values[:15] / len(results), 
           color='steelblue', edgecolor='black', alpha=0.7)
    plt.xlabel('Number of Flips', fontsize=12)
    plt.ylabel('Probability', fontsize=12)
    plt.title('Distribution: Flips Until First Heads', fontsize=14, weight='bold')
    plt.axvline(x=2, color='red', linestyle='--', linewidth=2, label='E[X] = 2')
    plt.legend()
    plt.grid(axis='y', alpha=0.3)
    
    plt.subplot(1, 2, 2)
    cumulative_avg = np.cumsum(results) / np.arange(1, len(results) + 1)
    plt.plot(cumulative_avg, linewidth=2, alpha=0.7, color='steelblue')
    plt.axhline(y=2, color='red', linestyle='--', linewidth=2, label='E[X] = 2')
    plt.xlabel('Number of Trials', fontsize=12)
    plt.ylabel('Cumulative Average', fontsize=12)
    plt.title('Convergence to Expected Value', fontsize=14, weight='bold')
    plt.legend()
    plt.grid(alpha=0.3)
    plt.xlim(0, 10000)
    
    plt.tight_layout()
    plt.show()

first_heads_recursive()
```

### 🎯 More Complex: Two Heads in a Row

```python
def two_heads_recursive():
    """Expected flips for TWO heads in a row"""
    
    print("\n\n🪙🪙 TWO HEADS IN A ROW")
    print("=" * 60)
    
    print("\nQuestion: Expected flips to get HH?")
    
    print("\n🧠 KEY INSIGHT: We need to track STATES")
    print("   State 0: Starting (no heads yet)")
    print("   State 1: Just got one head")
    print("   State 2: Got two heads (done!)")
    
    print("\n📐 DEFINE VARIABLES:")
    print("   E₀ = expected flips from State 0 (start)")
    print("   E₁ = expected flips from State 1 (one head)")
    
    print("\n🔄 FROM STATE 0:")
    print("   Flip once, then:")
    print("   • Heads (1/2): Move to State 1, need E₁ more")
    print("   • Tails (1/2): Stay in State 0, need E₀ more")
    print("   ")
    print("   E₀ = 1 + (1/2)×E₁ + (1/2)×E₀")
    
    print("\n🔄 FROM STATE 1:")
    print("   Flip once, then:")
    print("   • Heads (1/2): Done! 0 more flips")
    print("   • Tails (1/2): Back to State 0, need E₀ more")
    print("   ")
    print("   E₁ = 1 + (1/2)×0 + (1/2)×E₀")
    
    print("\n🔧 SOLVE THE SYSTEM:")
    print("   From equation 2: E₁ = 1 + E₀/2")
    print("   ")
    print("   Substitute into equation 1:")
    print("   E₀ = 1 + (1/2)(1 + E₀/2) + (1/2)E₀")
    print("   E₀ = 1 + 1/2 + E₀/4 + E₀/2")
    print("   E₀ = 3/2 + 3E₀/4")
    print("   E₀ - 3E₀/4 = 3/2")
    print("   E₀/4 = 3/2")
    print("   E₀ = 6")
    
    print("\n✅ ANSWER: Expected 6 flips to get two heads in a row")
    
    # State diagram visualization
    fig, ax = plt.subplots(figsize=(14, 8))
    
    # States
    states = {
        'Start': (0.2, 0.5),
        'One H': (0.5, 0.5),
        'Two H': (0.8, 0.5)
    }
    
    # Draw states
    for state, (x, y) in states.items():
        color = 'lightgreen' if state == 'Two H' else 'lightblue'
        circle = plt.Circle((x, y), 0.08, color=color, ec='black', linewidth=3)
        ax.add_patch(circle)
        ax.text(x, y, state, ha='center', va='center', fontsize=13, weight='bold')
    
    # Transitions
    # Start -> Start (Tails)
    ax.annotate('', xy=(0.15, 0.57), xytext=(0.15, 0.43),
               arrowprops=dict(arrowstyle='->', lw=2, color='red'))
    ax.text(0.08, 0.5, 'T\n(1/2)', ha='center', fontsize=10, weight='bold')
    
    # Start -> One H (Heads)
    ax.annotate('', xy=(0.42, 0.5), xytext=(0.28, 0.5),
               arrowprops=dict(arrowstyle='->', lw=3, color='green'))
    ax.text(0.35, 0.55, 'H (1/2)', ha='center', fontsize=10, weight='bold', color='green')
    
    # One H -> Start (Tails)
    ax.annotate('', xy=(0.28, 0.45), xytext=(0.42, 0.45),
               arrowprops=dict(arrowstyle='->', lw=2, color='red'))
    ax.text(0.35, 0.4, 'T (1/2)', ha='center', fontsize=10, weight='bold', color='red')
    
    # One H -> Two H (Heads)
    ax.annotate('', xy=(0.72, 0.5), xytext=(0.58, 0.5),
               arrowprops=dict(arrowstyle='->', lw=3, color='green'))
    ax.text(0.65, 0.55, 'H (1/2)', ha='center', fontsize=10, weight='bold', color='green')
    
    ax.set_xlim(0, 1)
    ax.set_ylim(0.2, 0.8)
    ax.axis('off')
    ax.set_title('State Diagram: Getting Two Heads in a Row', fontsize=16, weight='bold')
    
    plt.tight_layout()
    plt.show()
    
    # Verify with simulation
    def simulate_two_heads(trials=100000):
        results = []
        for _ in range(trials):
            flips = 0
            consecutive = 0
            while consecutive < 2:
                flips += 1
                if random.choice(['H', 'T']) == 'H':
                    consecutive += 1
                else:
                    consecutive = 0
            results.append(flips)
        return results
    
    print("\n🎲 SIMULATION (100,000 trials):")
    results = simulate_two_heads(100000)
    sim_mean = np.mean(results)
    print(f"   Average flips: {sim_mean:.4f}")
    print(f"   Theoretical: 6.0000")
    print(f"   Difference: {abs(sim_mean - 6):.4f} ✓")

two_heads_recursive()
```

### 🎓 The General Recursive Framework

```python
class RecursiveProbabilityProblem:
    """Framework for solving recursive probability problems"""
    
    def __init__(self, name):
        self.name = name
        self.states = {}
        self.transitions = {}
    
    def add_state(self, state_name):
        """Add a state to the problem"""
        self.states[state_name] = None
        self.transitions[state_name] = []
    
    def add_transition(self, from_state, to_state, probability, cost=1):
        """Add a transition between states"""
        self.transitions[from_state].append({
            'to': to_state,
            'prob': probability,
            'cost': cost
        })
    
    def describe(self):
        """Describe the recursive structure"""
        print(f"\n📋 RECURSIVE PROBLEM: {self.name}")
        print("=" * 60)
        
        for state in self.states:
            print(f"\nFrom {state}:")
            for trans in self.transitions[state]:
                print(f"  → {trans['to']} (prob={trans['prob']:.2f}, cost={trans['cost']})")
    
    def show_equations(self):
        """Show the recursive equations"""
        print(f"\n📐 RECURSIVE EQUATIONS:")
        print("=" * 60)
        
        for state in self.states:
            if state == 'DONE':
                continue
            
            equation_parts = [f"E[{state}] = {1}"]  # Cost of first action
            
            for trans in self.transitions[state]:
                if trans['to'] == 'DONE':
                    equation_parts.append(f" + ({trans['prob']:.2f})×0")
                else:
                    equation_parts.append(f" + ({trans['prob']:.2f})×E[{trans['to']}]")
            
            print(''.join(equation_parts))

# Example: Two heads in a row
print("\n🔧 FRAMEWORK EXAMPLE: Two Heads in a Row")

problem = RecursiveProbabilityProblem("Two Heads in a Row")
problem.add_state("Start")
problem.add_state("One H")
problem.add_state("DONE")

problem.add_transition("Start", "Start", 0.5)  # Tails
problem.add_transition("Start", "One H", 0.5)  # Heads
problem.add_transition("One H", "Start", 0.5)  # Tails
problem.add_transition("One H", "DONE", 0.5)   # Heads

problem.describe()
problem.show_equations()
```

### 🏆 Classic Application: Gambler's Ruin

```python
def gamblers_ruin_explanation():
    """The famous Gambler's Ruin problem"""
    
    print("\n\n🎰 GAMBLER'S RUIN")
    print("=" * 60)
    
    print("""
Scenario:
  • You have $5
  • Casino has $10
  • Each round: win $1 (prob 0.4) or lose $1 (prob 0.6)
  • Game ends when someone reaches $0 (ruined) or $15 (wins all)
  
Question: What's the probability you win everything?
    
This is recursive because:
  • From $5, you either go to $6 or $4
  • From $6, you either go to $7 or $5
  • ...and so on
  
Each state's winning probability depends on neighboring states!
    """)
    
    print("📐 Let P(k) = probability of winning starting with $k")
    print("\nBoundary conditions:")
    print("  P(0) = 0 (already ruined)")
    print("  P(15) = 1 (already won)")
    print("\nRecursive relation:")
    print("  P(k) = 0.4 × P(k+1) + 0.6 × P(k-1)")
    print("        win $1            lose $1")
    
    print("\n💡 This gives us 14 equations with 14 unknowns!")
    print("   Solvable, but tedious by hand...")
    
    # Solve numerically
    n_states = 16  # $0 to $15
    p_win_each_round = 0.4
    
    # Set up system of equations (simplified approach)
    probabilities = np.zeros(n_states)
    probabilities[15] = 1  # Boundary: already won
    
    # Iterate until convergence
    for _ in range(1000):
        new_prob = probabilities.copy()
        for k in range(1, n_states - 1):
            new_prob[k] = p_win_each_round * probabilities[k+1] + (1 - p_win_each_round) * probabilities[k-1]
        probabilities = new_prob
    
    print(f"\n✅ SOLUTION:")
    print(f"   P(win starting with $5) = {probabilities[5]:.4f} = {probabilities[5]*100:.2f}%")
    print(f"\n💡 Only {probabilities[5]*100:.0f}% chance of winning!")
    print(f"   The house edge (60% vs 40%) is devastating over time.")
    
    # Visualize
    plt.figure(figsize=(12, 6))
    plt.plot(range(n_states), probabilities, 'o-', linewidth=2, markersize=8, color='steelblue')
    plt.axvline(x=5, color='red', linestyle='--', linewidth=2, alpha=0.7, label='Your start ($5)')
    plt.axhline(y=probabilities[5], color='red', linestyle='--', linewidth=2, alpha=0.7)
    plt.xlabel('Money ($)', fontsize=12, weight='bold')
    plt.ylabel('P(Win Everything)', fontsize=12, weight='bold')
    plt.title('Gambler\'s Ruin: Probability of Winning vs Starting Money', fontsize=14, weight='bold')
    plt.legend(fontsize=11)
    plt.grid(alpha=0.3)
    plt.tight_layout()
    plt.show()

gamblers_ruin_explanation()
```

### ✅ Check Your Understanding

- [ ] Do you understand when recursion is useful?
- [ ] Can you set up a recursive equation?
- [ ] Can you identify the base cases (boundaries)?

```python
# 🏋️ PRACTICE PROBLEM

"""
Expected die rolls until you get a 6?

Set up the recursive equation and solve.
"""

# SOLUTION
print("\n\n🎲 DIE ROLLS UNTIL SIX")
print("=" * 60)

print("\nLet E = expected rolls to get a 6")
print("\nFirst roll:")
print("  • Roll 6 (prob 1/6): Done! 0 more rolls")
print("  • Roll 1-5 (prob 5/6): Back to start, need E more rolls")

print("\n📐 RECURSIVE EQUATION:")
print("  E = 1 + (1/6)×0 + (5/6)×E")

print("\n🔧 SOLVE:")
print("  E = 1 + 5E/6")
print("  E - 5E/6 = 1")
print("  E/6 = 1")
print("  E = 6")

print("\n✅ Expected 6 rolls to get a six!")

print("\n💡 GENERAL PATTERN:")
print("  For event with probability p:")
print("  Expected trials until first success = 1/p")
print("  ")
print("  Die (p=1/6): E = 6 ✓")
print("  Coin (p=1/2): E = 2 ✓")
```

---

# Final Section: Putting It All Together

## 🎓 Your Probability Journey

```python
print("🎓 CONGRATULATIONS! YOU'VE COMPLETED THE JOURNEY")
print("=" * 70)

topics_mastered = [
    "✅ Events & Sample Spaces - Understanding probability fundamentals",
    "✅ Three Axioms - The mathematical foundation",
    "✅ Set Operations - Combining events (AND, OR, NOT)",
    "✅ Counting Methods - Handling huge sample spaces",
    "✅ Conditional Probability - Updating with information",
    "✅ Independence - When events don't affect each other",
    "✅ Random Variables - Turning randomness into numbers",
    "✅ Expected Value - The long-run average",
    "✅ Linearity of Expectation - A powerful shortcut",
    "✅ Continuous Probability - From counts to measurements",
    "✅ Law of Total Probability - Breaking complex problems apart",
    "✅ Bayes' Rule - The crown jewel of learning",
    "✅ Recursion - When answers depend on themselves"
]

print("\n📚 TOPICS YOU'VE MASTERED:")
for topic in topics_mastered:
    print(f"  {topic}")

print("\n\n🎯 WHAT YOU CAN DO NOW:")
skills = [
    "Calculate probabilities from first principles",
    "Use set operations to combine events",
    "Count outcomes using permutations and combinations",
    "Update beliefs with Bayes' rule",
    "Calculate expected values and make better decisions",
    "Work with both discrete and continuous distributions",
    "Solve recursive probability problems",
    "Think probabilistically about uncertainty"
]

for skill in skills:
    print(f"  • {skill}")

print("\n\n💪 NEXT STEPS:")
next_steps = [
    "1. Practice daily with probability problems",
    "2. Apply probability to real-world decisions",
    "3. Study advanced topics:",
    "   • Markov chains (systems that transition between states)",
    "   • Statistical inference (learning from data)",
    "   • Machine learning (AI is applied probability)",
    "   • Stochastic processes (randomness over time)",
    "4. Teach someone else (best way to solidify understanding)",
    "5. Build projects that use probability"
]

for step in next_steps:
    print(f"  {step}")

print("\n\n🌟 REMEMBER:")
print("  The concepts are in your head RIGHT NOW.")
print("  Whether they stay there depends on what you do next.")
print("  Practice transforms understanding into mastery.")
print("\n  You're not just learning probability—")
print("  You're learning to think clearly about an uncertain world.")

print("\n\n" + "=" * 70)
print("                    YOU'RE NOW A PROBABILITY MASTER!")
print("=" * 70)
```

---

## 📝 Quick Reference Sheet

```python
"""
⚡ PROBABILITY QUICK REFERENCE SHEET
====================================

BASIC FORMULAS:
--------------
P(E) = favorable outcomes / total outcomes
P(not A) = 1 - P(A)
P(A or B) = P(A) + P(B) - P(A and B)
P(A and B) = P(A) × P(B)  [if independent]

CONDITIONAL:
-----------
P(A|B) = P(A and B) / P(B)
P(A and B) = P(A|B) × P(B) = P(B|A) × P(A)

BAYES' RULE:
-----------
P(A|B) = P(B|A) × P(A) / P(B)

LAW OF TOTAL PROBABILITY:
-------------------------
P(B) = Σ P(B|Aᵢ) × P(Aᵢ)

COUNTING:
--------
Permutations (order matters): P(n,k) = n!/(n-k)!
Combinations (order doesn't): C(n,k) = n!/[k!(n-k)!]
With repetition: n^k

EXPECTED VALUE:
--------------
E[X] = Σ x × P(X=x)
E[aX + bY] = a×E[X] + b×E[Y]  [always!]
E[XY] = E[X] × E[Y]  [if independent]

INDEPENDENCE TEST:
-----------------
A and B independent if:
  P(A and B) = P(A) × P(B)
  OR P(A|B) = P(A)
  OR P(B|A) = P(B)

DECISION TREE FOR COUNTING:
---------------------------
Can repeat? 
  YES → Order matters? → YES: n^k, NO: complex
  NO  → Order matters? → YES: P(n,k), NO: C(n,k)

CONTINUOUS:
----------
P(X = exact value) = 0
P(a ≤ X ≤ b) = ∫[a to b] f(x) dx
For Uniform(a,b): E[X] = (a+b)/2
"""

print(__doc__)
```

---

**You've done it! You now have a deep, intuitive understanding of probability. Keep practicing, keep applying it, and watch as you see the world through a probabilistic lens. Good luck on your journey!** 🎉
````
