Multi-Armed Bandits: A Deep Dive into Action-Value Methods, Non-Stationary Problems, and Exploration Strategies
The multi-armed bandit (MAB) problem is a classic reinforcement learning scenario that exemplifies the exploration-exploitation dilemma. Imagine yourself in a casino, faced with a row of slot machines, each with an unknown payout probability. Your goal is to maximize your winnings by strategically choosing which machines to play. This is the essence of the MAB problem, where each machine is an "arm" and the payout is the "reward." The core challenge lies in finding the right balance between "exploration" (trying different machines to learn their payout probabilities) and "exploitation" (playing the machine that seems to have the highest payout based on current knowledge). This trade-off is crucial for maximizing long-term rewards1.
This article delves into the intricacies of MAB, focusing on action-value methods, strategies for handling non-stationary problems, and the impact of initial values and upper confidence bound techniques.
Exploration vs. Exploitation
The exploration-exploitation dilemma is central to the MAB problem and reinforcement learning in general1. It arises from the need to balance two competing objectives:
Exploration: Trying out different actions (arms) to gather information about their potential rewards. This is essential for discovering new and potentially better actions.
Exploitation: Choosing the action that currently seems to offer the highest reward based on the available information. This aims to maximize immediate rewards.
Finding the optimal balance between exploration and exploitation is crucial for maximizing cumulative rewards over time. Too much exploration can lead to wasted resources on suboptimal actions, while too much exploitation can prevent the discovery of better actions.
Action-Value Methods
Action-value methods are a cornerstone of MAB solutions. They involve estimating the value of each action (arm) and using these estimates to guide decision-making. The core idea is to learn which actions yield the highest rewards over time.
Estimating Action Values
One of the simplest and most effective ways to estimate action values is through sample averaging. This involves calculating the average reward received each time an action is chosen. Formally, the value estimate Qₜ(a) for action 'a' at time step 't' is:
Qₜ(a) = (Sum of rewards when action 'a' was taken) / (Number of times action 'a' was taken)
As we accumulate more data, this estimate converges towards the true value of the action.
Updating Action Values Efficiently
While sample averaging provides a reliable estimate, it can be computationally expensive to store all past rewards and recompute the average at each step. A more efficient approach is to update the estimate incrementally. This involves updating the estimate based on the previous estimate and the new reward received2. The incremental update rule is:
Qₙ₊₁ = Qₙ + (1/n) * (Rₙ - Qₙ)
where:
Qₙ₊₁ is the updated estimate
Qₙ is the previous estimate
Rₙ is the reward received at the current step
n is the number of times the action has been taken
This rule allows us to update the estimate without storing all past rewards, making it computationally and memory-efficient.
Action Selection Strategies
Once we have the estimated action values, we need a strategy to select the next action. Here are a few common strategies:

Strategy
Description
Advantages
Disadvantages
Greedy
Always choose the action with the highest estimated value.
Simple to implement; exploits current knowledge.
Can get stuck in local optima; no exploration.
Epsilon-Greedy
Choose the greedy action most of the time, but with a small probability (epsilon), choose a random action.
Balances exploration and exploitation.
Can still lead to suboptimal performance in the long run due to continuous exploration, even when the optimal action has been identified2.

Select actions based on both their estimated rewards and the uncertainty associated with those estimates.

Epsilon-greedy is a widely used strategy that balances exploration and exploitation by introducing a small probability (epsilon) of choosing a random action6. However, it's important to note that this continuous exploration can sometimes lead to suboptimal performance in the long run, even when the optimal action has been identified2.
Thompson sampling, also known as probability matching, is another approach where a probability distribution is maintained for each action. Actions are then chosen based on samples drawn from these distributions. This can be more effective than epsilon-greedy in some scenarios6.
Evaluating Bandit Algorithms: Regret
Regret is a crucial concept in evaluating the performance of MAB algorithms8. It measures the difference between the cumulative reward obtained by the algorithm and the cumulative reward that would have been obtained by always choosing the optimal action. In other words, regret quantifies the "loss" incurred due to not knowing the optimal action beforehand.
Best Arm Identification
In some MAB problems, the goal is not just to maximize cumulative rewards but to identify the best arm with a certain level of confidence. This is known as best arm identification (BAI)1. There are two common settings for BAI:
Fixed Budget Setting: Given a fixed number of trials, the objective is to identify the best arm with the highest probability.
Fixed Confidence Setting: Given a desired confidence level, the objective is to identify the best arm with the minimum number of trials.
These settings reflect different priorities in BAI problems, where one might prioritize minimizing the number of trials or maximizing the probability of correctly identifying the best arm.
Tracking Non-Stationary Problems
In many real-world scenarios, the reward distributions of the arms are not fixed but change over time9. These are known as non-stationary problems. For example, in online advertising, the click-through rates of ads can change due to various factors like user preferences, time of day, and current events.
Challenges in Non-Stationary Environments
Traditional MAB algorithms, designed for stationary environments, may struggle to adapt to these dynamic changes. They might get stuck exploiting an arm that was initially good but has become suboptimal, or they might over-explore and miss out on exploiting the current best arm.
Adapting to Non-Stationary Problems
One way to handle non-stationary problems is to give more weight to recent rewards than to older ones. This can be achieved by using a constant step-size parameter (alpha) in the incremental update rule:
Qₙ₊₁ = Qₙ + alpha * (Rₙ - Qₙ)
where alpha is a value between 0 and 1. This gives more weight to recent rewards, allowing the agent to adapt to changes in the reward distributions.
Another approach is to use adaptive windowing techniques9. These techniques adjust the window size based on the detected changes in the environment. For example, if the algorithm detects a sudden shift in the reward distribution, it might shrink the window size to focus on the most recent data. Conversely, if the environment seems stable, it might expand the window size to incorporate more historical data.
Restless Bandits
A generalization of the non-stationary MAB problem is the "restless bandit problem," where the states of non-played arms can also evolve over time1. This adds another layer of complexity, as the agent needs to consider not only the changes in the reward distributions of the played arms but also the potential changes in the unplayed arms.
Effect of Setting Initial Values
The initial values assigned to the actions can significantly impact the agent's performance, especially in the early stages of learning.
Optimistic Initial Values
One effective technique is to use optimistic initial values. This involves setting the initial action values to be higher than what is realistically expected10. This encourages the agent to explore more in the beginning, as it is "optimistic" about the potential rewards of each action. As the agent explores and receives actual rewards, the estimates are adjusted downwards, leading to a better understanding of the true action values.
Impact on Exploration and Exploitation
Optimistic initial values promote exploration in the early stages, allowing the agent to quickly learn about the different actions10. This can lead to faster convergence to the optimal policy, especially in stationary environments. The higher the initial values, the more exploration the agent will perform10. This is because it takes longer for the inflated estimates to converge to the true action values, giving the agent more opportunities to explore different actions.
Using Upper Confidence Bound (UCB)
The Upper Confidence Bound (UCB) algorithm is another powerful approach to balance exploration and exploitation. It selects actions based on both their estimated rewards and the uncertainty associated with those estimates.
The UCB Formula
The UCB formula for selecting an action 'a' at time step 't' is:
UCB(a) = Qₜ(a) + c * sqrt(ln(t) / Nₜ(a))
where:
Qₜ(a) is the estimated value of action 'a'
Nₜ(a) is the number of times action 'a' has been selected
c is a confidence value that controls the level of exploration
t is the current time step
Balancing Exploration and Exploitation with UCB
The UCB algorithm balances exploration and exploitation by considering both the estimated reward and the uncertainty5. Actions that have been selected less often have higher uncertainty, and the UCB formula gives them a "boost," encouraging exploration. As an action is selected more frequently, the uncertainty decreases, and the algorithm focuses more on exploitation.
To illustrate, imagine two actions: action A has been selected many times and has a high estimated reward, while action B has been selected only a few times and has a lower estimated reward. The UCB algorithm will initially favor action B due to its higher uncertainty, even though its estimated reward is lower. This allows the algorithm to explore action B further and potentially discover that it is actually better than action A.
The UCB algorithm has been shown to be more efficient in exploration compared to epsilon-greedy, potentially leading to better performance in the long run4.
Types of Multi-Armed Bandit Problems
The MAB problem can be formulated in various ways depending on the nature of the rewards and the availability of additional information. Here are a few common types:
Bernoulli Bandits: In this type, the reward for each arm is either 0 or 1, representing a binary outcome like a click or no click on an advertisement12.
Contextual Bandits: In this type, the choice of action depends on the context or side information available to the agent. For example, in online advertising, the context might include user demographics, browsing history, or the current webpage7.
These different types of MAB problems require different algorithms and strategies to effectively balance exploration and exploitation.
Conclusion
The multi-armed bandit problem provides a valuable framework for understanding the exploration-exploitation dilemma in reinforcement learning. Action-value methods, strategies for handling non-stationary problems, and techniques like optimistic initial values and UCB offer powerful tools for solving MAB problems and maximizing rewards in various applications.
MAB algorithms have found widespread use in real-world scenarios, including:
Online Advertising: Selecting the most effective ads to display to users to maximize click-through rates and revenue7.
Clinical Trials: Determining the most effective treatments for patients while minimizing risks and maximizing positive outcomes6.
Recommender Systems: Personalizing recommendations to users by learning their preferences and suggesting items they are most likely to engage with.
By carefully considering the different aspects of MAB problems, such as action-value estimation, action selection strategies, and adaptation to non-stationary environments, we can design agents that effectively learn and adapt to dynamic environments, leading to improved decision-making and optimized outcomes in various domains.

```python
import torch
import torch.distributions as dist
from torch.utils.tensorboard import SummaryWriter
import numpy as np
import matplotlib.pyplot as plt # for inline plots in notebooks (optional)

# Bandit Environment Class
class BanditEnvironment:
    def __init__(self, num_arms, true_reward_means, reward_std=1.0):
        """
        Initializes the bandit environment.

        Args:
            num_arms (int): Number of bandit arms.
            true_reward_means (list or numpy array): True mean reward for each arm.
            reward_std (float): Standard deviation of the reward distribution.
        """
        if len(true_reward_means) != num_arms:
            raise ValueError("Length of true_reward_means must equal num_arms")
        self.num_arms = num_arms
        self.true_reward_means = torch.tensor(true_reward_means, dtype=torch.float32)
        self.reward_std = reward_std

    def pull_arm(self, arm_index):
        """
        Simulates pulling a bandit arm and getting a reward.

        Args:
            arm_index (int): Index of the arm to pull (0-indexed).

        Returns:
            float: Reward obtained from pulling the arm.
        """
        if not 0 <= arm_index < self.num_arms:
            raise ValueError("Invalid arm_index")
        reward_distribution = dist.Normal(self.true_reward_means[arm_index], self.reward_std)
        reward = reward_distribution.sample()
        return reward.item()


# Agent Classes for Different Strategies

class EpsilonGreedyAgent:
    def __init__(self, num_arms, epsilon=0.1):
        """
        Epsilon-Greedy Agent.

        Args:
            num_arms (int): Number of bandit arms.
            epsilon (float): Exploration probability (0 to 1).
        """
        self.num_arms = num_arms
        self.epsilon = epsilon
        self.estimated_rewards = torch.zeros(num_arms, dtype=torch.float32) # Q-values, initialized to 0
        self.arm_pull_counts = torch.zeros(num_arms, dtype=torch.int)

    def select_action(self):
        """Selects an action (arm index) based on epsilon-greedy strategy."""
        if torch.rand(1) < self.epsilon: # Explore
            return torch.randint(0, self.num_arms, (1,)).item() # Random arm index
        else: # Exploit
            return torch.argmax(self.estimated_rewards).item() # Arm with highest estimated reward

    def update_estimates(self, arm_index, reward):
        """Updates the estimated reward for the chosen arm."""
        self.arm_pull_counts[arm_index] += 1
        n = self.arm_pull_counts[arm_index]
        # Incremental average update:  Q_new = Q_old + (1/n) * (reward - Q_old)
        self.estimated_rewards[arm_index] += (reward - self.estimated_rewards[arm_index]) / n


class UCBAgent:
    def __init__(self, num_arms, c=2):
        """
        Upper Confidence Bound (UCB) Agent.

        Args:
            num_arms (int): Number of bandit arms.
            c (float): Exploration-exploitation tradeoff parameter (higher c = more exploration).
        """
        self.num_arms = num_arms
        self.c = c
        self.estimated_rewards = torch.zeros(num_arms, dtype=torch.float32)
        self.arm_pull_counts = torch.zeros(num_arms, dtype=torch.int)
        self.time_step = 0 # For UCB formula

    def select_action(self):
        """Selects an action based on the UCB strategy."""
        self.time_step += 1
        ucb_values = torch.zeros(self.num_arms, dtype=torch.float32)
        for arm_index in range(self.num_arms):
            if self.arm_pull_counts[arm_index] == 0: # Explore arms that haven't been pulled yet
                return arm_index
            else:
                confidence_bound = self.c * torch.sqrt(torch.log(torch.tensor(self.time_step)) / self.arm_pull_counts[arm_index])
                ucb_values[arm_index] = self.estimated_rewards[arm_index] + confidence_bound
        return torch.argmax(ucb_values).item()

    def update_estimates(self, arm_index, reward):
        """Updates the estimated reward for the chosen arm."""
        self.arm_pull_counts[arm_index] += 1
        n = self.arm_pull_counts[arm_index]
        self.estimated_rewards[arm_index] += (reward - self.estimated_rewards[arm_index]) / n


class ThompsonSamplingAgent:
    def __init__(self, num_arms):
        """
        Thompson Sampling Agent (Gaussian prior and likelihood).

        Args:
            num_arms (int): Number of bandit arms.
        """
        self.num_arms = num_arms
        # Prior distributions for each arm (Gaussian, initialized with mean 0, variance 1)
        self.prior_means = torch.zeros(num_arms, dtype=torch.float32)
        self.prior_variances = torch.ones(num_arms, dtype=torch.float32)

    def select_action(self):
        """Selects an action based on Thompson Sampling."""
        sampled_rewards = torch.zeros(self.num_arms, dtype=torch.float32)
        for arm_index in range(self.num_arms):
            # Sample from the posterior distribution (using current prior as approx. posterior)
            sampled_rewards[arm_index] = dist.Normal(self.prior_means[arm_index], torch.sqrt(self.prior_variances[arm_index])).sample()
        return torch.argmax(sampled_rewards).item() # Choose arm with highest sampled reward

    def update_estimates(self, arm_index, reward):
        """Updates the prior distribution for the chosen arm (Bayesian update for Gaussian)."""
        # For simplicity, we assume known reward variance (can be extended to unknown variance)
        reward_variance = 1.0  # Assume reward variance is 1 (same as reward_std in environment)
        prior_variance = self.prior_variances[arm_index]
        prior_mean = self.prior_means[arm_index]

        # Update posterior parameters (Bayesian update for Gaussian-Gaussian)
        posterior_variance = 1.0 / (1.0 / prior_variance + 1.0 / reward_variance)
        posterior_mean = posterior_variance * (prior_mean / prior_variance + reward / reward_variance)

        self.prior_means[arm_index] = posterior_mean
        self.prior_variances[arm_index] = posterior_variance


# Experiment Runner Function
def run_experiment(agent, environment, num_steps, writer, strategy_name):
    """
    Runs an experiment for a given agent and environment.

    Args:
        agent: Agent object (EpsilonGreedyAgent, UCBAgent, ThompsonSamplingAgent).
        environment: BanditEnvironment object.
        num_steps (int): Number of steps to run the experiment.
        writer (SummaryWriter): TensorBoard SummaryWriter for logging.
        strategy_name (str): Name of the strategy for logging.

    Returns:
        list: Cumulative rewards over time.
    """
    cumulative_reward = 0
    cumulative_rewards = []

    for step in range(num_steps):
        action = agent.select_action()
        reward = environment.pull_arm(action)
        agent.update_estimates(action, reward)
        cumulative_reward += reward
        cumulative_rewards.append(cumulative_reward)

        writer.add_scalar(f'Cumulative Reward/{strategy_name}', cumulative_reward, step)

    return cumulative_rewards


# --- Main Experiment Setup ---
if __name__ == '__main__':
    num_arms = 4
    true_reward_means = [0.1, 0.3, 0.2, 0.7] # Example true means (arm 3 is the best)
    num_steps = 2000
    num_experiments = 20 # Run multiple experiments to average out randomness

    # Initialize Bandit Environment
    environment = BanditEnvironment(num_arms, true_reward_means)

    # Initialize Agents and TensorBoard Writers
    agents = {
        "EpsilonGreedy": EpsilonGreedyAgent(num_arms, epsilon=0.1),
        "UCB": UCBAgent(num_arms, num_arms), # c value can be tuned, num_arms or 2 are common starting points
        "ThompsonSampling": ThompsonSamplingAgent(num_arms)
    }
    writers = {
        name: SummaryWriter(log_dir=f'runs/mab_strategies/{name}') for name in agents
    }

    avg_cumulative_rewards = {name: np.zeros(num_steps) for name in agents}

    # Run Experiments and Average Results
    for experiment_index in range(num_experiments):
        print(f"Experiment {experiment_index + 1}/{num_experiments}")
        for strategy_name, agent in agents.items():
            writer = writers[strategy_name] # Get the appropriate writer for this strategy
            cumulative_rewards = run_experiment(agent, environment, num_steps, writer, strategy_name)
            avg_cumulative_rewards[strategy_name] += np.array(cumulative_rewards)

            # Re-initialize agents for the next experiment (important!)
            if strategy_name == "EpsilonGreedy":
                agents[strategy_name] = EpsilonGreedyAgent(num_arms, epsilon=0.1)
            elif strategy_name == "UCB":
                agents[strategy_name] = UCBAgent(num_arms, num_arms)
            elif strategy_name == "ThompsonSampling":
                agents[strategy_name] = ThompsonSamplingAgent(num_arms)


    # Final Averaging and (Optional) Matplotlib Plotting
    for strategy_name in agents:
        avg_cumulative_rewards[strategy_name] /= num_experiments

    print("Experiments finished. Run TensorBoard to visualize results:")
    print("tensorboard --logdir runs/mab_strategies")

    # --- Optional: Matplotlib Plotting for Notebooks ---
    # if 'ipykernel' in sys.modules: # Check if running in Jupyter/IPython
    #     plt.figure(figsize=(10, 6))
    #     for strategy_name, rewards in avg_cumulative_rewards.items():
    #         plt.plot(rewards, label=strategy_name)
    #     plt.xlabel("Steps")
    #     plt.ylabel("Average Cumulative Reward")
    #     plt.title("Average Cumulative Reward Comparison")
    #     plt.legend()
    #     plt.grid(True)
    #     plt.show()
```
